# 🧠  What is a Bernoulli Trial?

Before understanding the Binomial Distribution, we need to understand:

# 🪙 Bernoulli Trial

A Bernoulli Trial is a random experiment that has exactly **two possible outcomes**.

These outcomes are usually called:

```text
Success
Failure
```

For example:

### 🪙 Coin Toss

```text
Heads → Success
Tails → Failure
```

### 🎯 Quality Check

```text
Defective → Success
Not Defective → Failure
```

### 📧 Email Classification

```text
Spam → Success
Not Spam → Failure
```

The labels "Success" and "Failure" are just names.

They do not necessarily mean:

```text
Success = Good
Failure = Bad
```

They simply represent the two possible outcomes.

---


# 🎯  Examples of Bernoulli Trials

| Experiment | Success | Failure |
|---|---|---|
| 🪙 Coin Toss | Heads | Tails |
| 🎯 Target Practice | Hit | Miss |
| 🏭 Quality Check | Defective | Not Defective |
| 📧 Email | Spam | Not Spam |
| 🛒 Purchase | Purchased | Not Purchased |
| 🧪 Experiment | Successful | Unsuccessful |
| 🤖 Prediction | Correct | Incorrect |

Each individual trial has:

```text
Exactly 2 possible outcomes
```

---

# 🧠 Bernoulli Random Variable

We can represent the outcomes numerically.

Usually:

```text
Success = 1
Failure = 0
```

Therefore:

```text
X ∈ {0, 1}
```

For example:

```text
X = 1 → Success
X = 0 → Failure
```

If the probability of success is:

```text
p
```

then the probability of failure is:

```text

```

We can represent this as:

```text
P(X = 1) = p
```

and:

```text
P(X = 0) = 1 - p
```

# 🧠  What is a Binomial Distribution?

A Binomial Distribution models the number of successes in a fixed number of independent Bernoulli trials.

In simple words:

> **Repeat a two-outcome experiment a fixed number of times and count the number of successes.**

For example:

```text
10 coin tosses
        ↓
Count Heads
        ↓
Number of Heads
```

The number of Heads can be:

```text
0, 1, 2, 3, ..., 10
```

This is a Binomial Random Variable.

---

# 🎯 Binomial Random Variable

We usually represent the number of successes as:

```text
X
```

If there are:

```text
n = Number of trials
```

and:

```text
p = Probability of success
```

then:

```text
X ~ Binomial(n, p)
```

For example:

```text
X ~ Binomial(10, 0.5)
```

means:

> X represents the number of successes in 10 independent trials where the probability of success in each trial is 0.5.

---

# 🔢  Conditions for Binomial Distribution

A situation can be modeled using a Binomial Distribution when the following conditions are satisfied.

## 1️⃣ Fixed Number of Trials

There must be a fixed number of trials:

```text
n
```

For example:

```text
10 coin tosses
```

---
## 2️⃣ Two Possible Outcomes

Each trial must have exactly two outcomes:

```text
Success
Failure
```

---

## 3️⃣ Independent Trials

The outcome of one trial should not affect another trial.

For example, in repeated fair coin tosses:

```text
Previous toss
      ↓
does not affect
      ↓
Next toss
```

---
## 4️⃣ Constant Probability of Success

The probability of success should remain the same for every trial.

For example:

```text
P(Success) = 0.5
```

for every coin toss.

---

## 5️⃣ Count the Number of Successes

The random variable represents:

```text
Number of Successes
```

---


# 🧠 Remember the Five Conditions

A Binomial Distribution requires:

```text
F → Fixed number of trials
I → Independent trials
T → Two outcomes
C → Constant probability
S → Count successes
```

You can remember:

# ⭐ FITCS

```text
F → Fixed
I → Independent
T → Two outcomes
C → Constant probability
S → Successes counted
```

# 🧮  Binomial Probability Formula

The probability of getting exactly `k` successes in `n` trials is:

```text
P(X = k)
=
C(n, k) p^k (1-p)^(n-k)
```

Where:

```text
n = Number of trials

k = Number of successes

p = Probability of success

1-p = Probability of failure

C(n,k) = Number of ways to arrange k successes among n trials
```

---

# 🔢  Understanding C(n, k)

The term:

```text
C(n, k)
```

is called:

```text
Combination
```

It is calculated as:

```text
C(n,k) = n! / [k!(n-k)!]
```

It tells us:

> How many different ways we can arrange `k` successes among `n` trials.

---

# 🧠  Why Do We Need Combinations?

Suppose we toss a coin 3 times and want exactly 2 Heads.

Possible sequences include:

```text
HHT
HTH
THH
```

There are:

```text
3
```

different arrangements.

Therefore:

```text
C(3,2) = 3
```

The combination term accounts for all possible arrangements of the successes.

---

# 🧮  Example — Exactly 2 Heads in 3 Tosses

Given:

```text
n = 3
k = 2
p = 0.5
```

Formula:

```text
P(X = 2)
=
C(3,2)(0.5)^2(0.5)^1
```

Since:

```text
C(3,2) = 3
```

we get:

```text
P(X = 2)
=
3 × 0.25 × 0.5
```

```text
= 0.375
```

Therefore:

```text
P(X = 2) = 0.375
```

or:

```text
37.5%
```

---


```text

```

Exactly 2 Heads occur in:

```text

```

Therefore:

```text

=

```

```text

```

This matches the Binomial Formula.

# 🎯 Another Example — Customer Purchases

Suppose:

```text
Probability of purchase = 0.3
```

We observe:

```text
10 customers
```

We want to know:

> What is the probability that exactly 4 customers purchase the product?

Define:

```text
n = 10
p = 0.3
k = 4
```

Then:

```text
P(X = 4)
=
C(10,4)(0.3)^4(0.7)^6
```

---

# 🧮  Calculate Manually with Python

````python
import math

n = 10
k = 4
p = 0.3

combination = math.comb(n, k)

probability = (
    combination
    * (p ** k)
    * ((1 - p) ** (n - k))
)

print("Combination:", combination)
print("Probability:", probability)
````

# 📊 Understanding the Result

The result represents:

```text
P(X = 4)
```

In simple words:

> The probability that exactly 4 out of 10 customers make a purchase when each customer has a 30% probability of purchasing.

# 📊  PMF of a Binomial Distribution

The Binomial Distribution is a **discrete probability distribution**.

Therefore, it has a:

The PMF gives:

# 🐍  Binomial Distribution with SciPy

SciPy provides the Binomial distribution through:


````python
from scipy.stats import binom

n = 10
p = 0.5

distribution = binom(
    n=n,
    p=p
)
````

# 🎯 Probability of Exactly k Successes

Suppose we want:

```text
P(X = 6)
```

Use:


````python
probability = distribution.pmf(6)

print(
    "P(X = 6):",
    probability
)
````

# 📊  Generate the Complete PMF


````python
import numpy as np

x = np.arange(
    0,
    n + 1
)

probabilities = distribution.pmf(x)

print(
    probabilities
)
````

# 📈  Visualizing the Binomial Distribution


````python
import matplotlib.pyplot as plt

plt.bar(
    x,
    probabilities
)

plt.title(
    "Binomial Distribution"
)

plt.xlabel(
    "Number of Successes"
)

plt.ylabel(
    "Probability"
)

plt.xticks(x)

plt.show()
````

### 👀 Observation

For:

```text

p = 0.5
```

the distribution is approximately symmetric around:

```text

```

because the expected number of successes is:


# 📏  Variance of a Binomial Distribution

The variance is:

```text
Variance = np(1-p)
```

For:

```text

```

we get:

```text
Variance
=
10 × 0.3 × 0.7
```

```text
= 2.1
```

---


# 📐  Standard Deviation

Standard deviation is the square root of variance:

```text
σ = √[np(1-p)]
```

For:

```text
n = 10
p = 0.3
```

we get:

```text
σ = √2.1
```

```text
≈ 1.449
```

---

# 🧠  Mean, Variance, and Standard Deviation Summary

For:

```text
X ~ Binomial(n,p)
```

### Mean

```text
μ = np
```

### Variance

```text
σ² = np(1-p)
```

### Standard Deviation

```text
σ = √[np(1-p)]
```

---

# 🐍  Calculate Mean and Variance Using Python


````python
n = 10
p = 0.3

mean = n * p

variance = (
    n * p * (1 - p)
)

std = np.sqrt(variance)

print("Mean:", mean)
print("Variance:", variance)
print("Standard Deviation:", std)
````

# 📊  Verify Using SciPy


````python
distribution = binom(
    n=10,
    p=0.3
)

print(
    "Mean:",
    distribution.mean()
)

print(
    "Variance:",
    distribution.var()
)

print(
    "Standard Deviation:",
    distribution.std()
)
````

# 📊 Binomial Distribution and Normal Approximation

For sufficiently large values of `n`, under suitable conditions, a Binomial Distribution can be approximated by a Normal Distribution.

A common rule of thumb checks:

```text
np ≥ 5
```

and:

```text
n(1-p) ≥ 5
```

Some applications use stricter thresholds such as 10.

The approximation improves as the expected numbers of successes and failures become sufficiently large.

---

# 🧠  Why Does This Matter?

This creates an important connection:

```text
Binomial Distribution
        ↓
Large n
        ↓
Approximation
        ↓
Normal Distribution
```

This is one reason why the Normal Distribution is so important in Statistics.

---

# 🎯 Example of Normal Approximation

Suppose:

```text
n = 100
p = 0.5
```

Then:

```text
np = 50
```

and:

```text
n(1-p) = 50
```

Both are sufficiently large.

Therefore, the Binomial Distribution may be reasonably approximated by a Normal Distribution.

---

# 🧠  Binomial Distribution vs Normal Distribution

| 🎯 Binomial | 🔔 Normal |
|---|---|
| Discrete | Continuous |
| Counts successes | Models continuous measurements |
| Fixed number of trials | Continuous range |
| Parameters `n, p` | Parameters `μ, σ` |
| Uses PMF | Uses PDF |
| `X = 0,1,2,...,n` | `X` can take values over a continuous range |
| Can approximate Normal for large `n` | Used as an approximation in many settings |

# 🎯 Probability of At Least k Successes

Suppose we want:



Using Python:

````python
probability = (
    1 -
    binom.cdf(
        4,
        n,
        p
    )
)

print(
    "P(X >= 5):",
    probability
)
````

# 🧠 Exactly vs At Most vs At Least


# 🧠  Probability Language

|---|---|

| Less than 5 | `X < 5` |
| More than 5 | `X > 5` |
| Between 3 and 7 inclusive | `3 ≤ X ≤ 7` |

Understanding these phrases is very important when solving probability problems.

# 🧠 Theoretical vs Experimental Probability

There are two ideas:

### 📐 Theoretical Probability

Calculated using:

```text
Binomial Formula
```

or:

```text
SciPy
```

### 🧪 Experimental Probability

Estimated from repeated simulations.

For example:

```text
Number of experiments where X = 5
────────────────────────────────
Total number of experiments
```

As the number of simulations increases, the experimental probability generally gets closer to the theoretical probability.

---

# 🧮  Compare Theoretical and Experimental Probability


````python
theoretical_probability = binom.pmf(
    5,
    10,
    0.5
)

experimental_probability = np.mean(
    results == 5
)

print(
    "Theoretical Probability:",
    theoretical_probability
)

print(
    "Experimental Probability:",
    experimental_probability
)
````

# ⚠️  Common Mistakes

### ❌ Mistake 1 — Using Binomial for More Than Two Outcomes

A Binomial Trial requires:

```text
Two outcomes
```

If an experiment has:

```text
Red
Blue
Green
Yellow
```

you need a different probability model unless the outcomes are grouped into two categories.

---

### ❌ Mistake 2 — Changing Probability Between Trials
The probability of success should remain constant.

For example:

```text
p = 0.5
```

for every trial.

---

### ❌ Mistake 3 — Ignoring Independence

Trials should be independent under the Binomial model.

If one outcome affects the next, the simple Binomial model may not be appropriate.

---

### ❌ Mistake 4 — Not Fixing the Number of Trials

The number of trials must be known:

```text
n
```

before calculating the Binomial probability.

---

### ❌ Mistake 5 — Confusing Bernoulli and Binomial

### Bernoulli:

```text
One trial
```

### Binomial:

```text
Multiple Bernoulli trials
+
Count successes
```

---

### ❌ Mistake 6 — Confusing Probability and Expected Value

If:

```text
Expected successes = 20
```

it does not mean:

```text
Probability = 20
```

The expected value represents an average number of successes.

---

# 🧠  Bernoulli vs Binomial

| 🪙 Bernoulli | 🎯 Binomial |
|---|---|
| One trial | Multiple trials |
| Two outcomes | Two outcomes per trial |
| `X ∈ {0,1}` | `X ∈ {0,1,...,n}` |
| Success probability `p` | Parameters `n, p` |
| Example: One coin toss | Example: 10 coin tosses |
| Models one outcome | Models number of successes |

---

# 🎯 Binomial Distribution vs Normal Distribution

| 🎯 Binomial | 🔔 Normal |
|---|---|
| Discrete | Continuous |
| Counts successes | Models continuous values |
| Fixed number of trials | No fixed trial count |
| Parameters `n, p` | Parameters `μ, σ` |
| Uses PMF | Uses PDF |
| `X = 0,...,n` | Continuous range |
| Can approximate Normal | Can approximate Binomial for large n |

## 🎯 Binomial Distribution

```text
X ~ Binomial(n, p)
```

---

## 📊 Probability of Exactly k Successes

```text
P(X = k)
=
C(n,k) p^k (1-p)^(n-k)
```

---

## 🔢 Combination

```text
C(n,k)
=
n!
────────────
k!(n-k)!
```

---

## 🎯 Probability of Failure

```text
q = 1 - p
```

---

## 🧠 Mean

```text
μ = np
```

---

## 📊 Variance

```text
σ² = np(1-p)
```

---

## 📏 Standard Deviation

```text
σ = √[np(1-p)]
```

# 💎  Key Insights

- 🪙 A **Bernoulli Trial** has exactly two possible outcomes.
- 🎯 A **Binomial Distribution** counts successes across multiple Bernoulli trials.
- 🔢 The number of trials is represented by `n`.
- 🎯 Probability of success is represented by `p`.
- 🔴 Probability of failure is `1-p`.
- 📊 The random variable represents the number of successes.
- 📐 The Binomial PMF calculates the probability of exactly `k` successes.
- 🧠 The mean is:

```text
np
```

- 📊 The variance is:

```text
np(1-p)
```

- 📏 The standard deviation is:

```text
√[np(1-p)]
```

- 🤖 Binary outcomes make Binomial modeling useful in many Data Science problems.
- 📈 For sufficiently large `n`, Binomial distributions can often be approximated by Normal distributions.
- ⚠️ Independence and constant probability are important assumptions.
