# 📐  What is Skewness?

Skewness measures the:

> **Asymmetry of a probability distribution.**

In simple words:

> **Skewness tells us whether the distribution is balanced or has a longer tail on one side.**

A distribution can be:

```text
↔️ Symmetric
```

```text
📈 Positively Skewed
```

or:

```text
📉 Negatively Skewed
```

---

# ⚖️  Symmetric Distribution

A distribution is approximately symmetric when:

```text
Left side ≈ Right side
```

Conceptually:

```text
             🔔
           /    \
         /        \
       /            \
______/              \______
```

The two sides have a similar shape.

For a perfectly symmetric distribution:

```text
Skewness = 0
```

---

# 🔔  Normal Distribution and Skewness

A Normal Distribution is symmetric.

Therefore:

```text
Skewness = 0
```

Conceptually:

```text
          /\
        /    \
      /        \
_____/          \_____
```

The left and right tails are balanced.

---

# 📈  Positive Skewness

Positive Skewness is also called:

```text
Right Skewness
```

because the distribution has a longer tail toward the:

```text
Right side
```

Conceptually:

```text
        🔔
       /  \
      /    \
_____/      \________________
```

The right tail extends farther.

---


# 🧠  Why is it Called Positive?

Because the longer tail extends toward:

```text
Positive direction
```

or:

```text
Right side
```

Therefore:

```text
Positive Skewness
=
Right Skewness
```

---

# 💰  Real-World Example of Positive Skewness

Income is a common example.

Suppose most people earn:

```text
₹20,000 – ₹50,000
```

while a small number of people earn:

```text
₹5,00,000+
```

Those very high values create a long right tail.

This creates:

```text
Positive Skewness
```

---

# 📉  Negative Skewness

Negative Skewness is also called:

```text
Left Skewness
```

because the longer tail extends toward the:

```text
Left side
```
The left tail extends farther.

 ---

# 🧠  Why is it Called Negative?

Because the longer tail extends toward the:

```text
Negative direction
```

or:

```text
Left side
```

Therefore:

```text
Negative Skewness
=
Left Skewness
```

---

# 🎓  Real-World Example of Negative Skewness

Suppose a very easy exam is given.

Most students score:

```text
80 – 100
```

while only a few students score:

```text
20 – 40
```

The lower scores create a longer left tail.

Therefore the distribution may be:

```text
Negatively Skewed
```

# 🧠 Important Relationship Between Mean, Median, and Mode

Skewness often affects the relationship between:

```text
Mean
Median
Mode
```

### Symmetric Distribution

Approximately:

```text
Mean ≈ Median ≈ Mode
```



### Positive Skewness

Typically:

```text
Mode < Median < Mean
```

The long right tail pulls the mean toward the right.



### Negative Skewness

Typically:

```text
Mean < Median < Mode
```

The long left tail pulls the mean toward the left.

---


# 📊 Mean-Median Relationship

Remember:

```text
Positive Skew
      ↓
Long Right Tail
      ↓
Mean pulled right
      ↓
Mean > Median
```

And:

```text
Negative Skew
      ↓
Long Left Tail
      ↓
Mean pulled left
      ↓
Mean < Median
```

This is an intuitive way to understand skewness.

---

# 🧮  Mathematical Intuition Behind Skewness

Skewness is related to the:

```text
Third Central Moment
```

The basic idea is:

```text
Skewness
≈
Third Central Moment
────────────────────
Standard Deviation³
```

One common population formula is:

```text
γ₁
=
E[(X - μ)³]
──────────────
σ³
```

Where:

```text
X = Observation

μ = Mean

σ = Standard Deviation

E = Expected Value
```

---



# 🧠  Why the Third Power?

Notice:

```text
(X - μ)³
```

The difference from the mean is cubed.

This is important because:

### Positive deviations

```text
(+)^3 = Positive
```

### Negative deviations

```text
(-)^3 = Negative
```

Therefore, the direction of the tail is preserved.

This helps Skewness detect asymmetry.

---


# 🎯  Understanding Positive Skew Mathematically

Suppose there are some extremely large values.

Then:

```text
X - μ
```

can be very large and positive.

Cubing makes them even larger:

```text
Large Positive Deviation
        ↓
Cubed
        ↓
Very Large Positive Contribution
```

Therefore:

```text
Positive Skewness
```

can result.

 ---


# 📉  Understanding Negative Skew Mathematically

Suppose there are some extremely small values.

Then:

```text
X - μ
```

can be large and negative.

After cubing:

```text
Large Negative Deviation
        ↓
Cubed
        ↓
Very Large Negative Contribution
```

Therefore:

```text
Negative Skewness
```

can result.

---


# 📊  Skewness Interpretation

A rough interpretation can be:

```text
Skewness ≈ 0
        ↓
Approximately Symmetric
```

```text
Skewness > 0
        ↓
Right Skewed
```

```text
Skewness < 0
        ↓
Left Skewed
```

However:

> There is no universal cutoff that perfectly classifies every dataset.

The histogram and domain context should also be considered.

# 🧠 What is Kurtosis?

Now let's understand:

# 📊 Kurtosis

Kurtosis describes aspects of the distribution's:

```text
Tail behavior
```

and its tendency to produce:

```text
Extreme observations
```

A common misconception is:

> "Kurtosis only tells us how peaked a distribution is."

This is an oversimplification.

A more useful interpretation is:

> **Kurtosis is strongly related to the heaviness of the tails and the tendency to produce extreme values.**

---

# 📐  Mathematical Intuition Behind Kurtosis

Kurtosis is related to the:

```text
Fourth Central Moment
```

A common population definition is:

```text
Kurtosis
=
E[(X - μ)⁴]
────────────
σ⁴
```

Where:

```text
X = Observation

μ = Mean

σ = Standard Deviation
```

Because deviations are raised to the fourth power:

```text
(X - μ)⁴
```

large deviations receive very large weights.

Therefore, Kurtosis is highly influenced by extreme observations.

---

# 🧠  Why the Fourth Power?

Consider:

```text
Deviation = 2
```

Then:

```text
2⁴ = 16
```

But if:

```text
Deviation = 5
```

then:

```text
5⁴ = 625
```

And:

```text
Deviation = 10
```

gives:

```text
10⁴ = 10,000
```

So extreme observations can have a very large influence on Kurtosis.

---

# 📊 Types of Kurtosis

There are three commonly discussed categories:

```text
             Kurtosis
                 │
       ┌─────────┼─────────┐
       ↓         ↓         ↓
   Platykurtic Mesokurtic Leptokurtic
       │         │         │
    Lighter    Normal    Heavier
     Tails      Like       Tails
```

Let's understand each one.

---

# 📉 Platykurtic Distribution

A Platykurtic distribution generally has:

```text
Lighter Tails
```

and fewer extreme observations compared with a Normal distribution.

Conceptually:

```text
       _________
     /           \
____/             \____
```

The distribution tends to have less extreme tail behavior.

---

# 🔔 Mesokurtic Distribution

A Mesokurtic distribution has kurtosis characteristics similar to the Normal Distribution.

For the Normal Distribution:

```text
Pearson Kurtosis = 3
```

and:

```text
Excess Kurtosis = 0
```

This distinction is important.

---

# 📈  Leptokurtic Distribution

A Leptokurtic distribution generally has:

```text
Heavier Tails
```

which means a greater tendency for extreme observations compared with a Normal distribution.

Conceptually:

```text
          /\
         /  \
        /    \
_______/      \_______
```

The key point is:

```text
Heavier Tails
```

rather than simply:

```text
Sharper Peak
```

---

# 🧠  Pearson Kurtosis vs Excess Kurtosis

This is a very important distinction.

There are two common ways to report Kurtosis.

### Pearson Kurtosis

For a Normal Distribution:

```text
Kurtosis = 3
```

### Excess Kurtosis

We subtract 3:

```text
Excess Kurtosis
=
Kurtosis - 3
```

Therefore, for a Normal Distribution:

```text
Excess Kurtosis = 0
```
 
 ---


# 📊  Kurtosis Interpretation

Using **Excess Kurtosis**:

```text
Excess Kurtosis = 0
        ↓
Mesokurtic
```

```text
Excess Kurtosis > 0
        ↓
Leptokurtic
        ↓
Heavier Tails
```

```text
Excess Kurtosis < 0
        ↓
Platykurtic
        ↓
Lighter Tails
```

---

# 🧠  Why Excess Kurtosis is Commonly Used

If we use Pearson Kurtosis:

```text
Normal = 3
```

which can make interpretation less intuitive.

Excess Kurtosis shifts the Normal baseline to:

```text
0
```

So:

```text
0 → Normal-like
Positive → Heavier tails
Negative → Lighter tails
```

This is commonly used in Python statistical tools.

---

# 📐  Skewness vs Kurtosis

These two measures describe different characteristics.

### 📐 Skewness

Focuses mainly on:

```text
Asymmetry
```

Question:

> Is the distribution pulled more toward one side?



### 📊 Kurtosis

Focuses mainly on:

```text
Tail behavior
```

Question:

> How prone is the distribution to extreme observations compared with a reference distribution?

---


# 📊  Comparison Table

| 📐 Skewness | 📊 Kurtosis |
|---|---|
| Measures asymmetry | Describes tail behavior |
| Uses third moment | Uses fourth moment |
| Positive → Right Skew | Positive Excess → Heavier Tails |
| Negative → Left Skew | Negative Excess → Lighter Tails |
| Zero → Symmetric | Zero Excess → Normal-like reference |
| Sensitive to asymmetry | Sensitive to extreme values |

---


# 🧠  Important Distinction

Do not memorize:

```text
Skewness = Shape
Kurtosis = Peak
```

Instead remember:

```text
Skewness
   ↓
Asymmetry
   ↓
Which side has the longer tail?
```

and:

```text
Kurtosis
   ↓
Tail Behavior
   ↓
How much extreme-value tendency?
```

---

# 🐍  Generate Normal Data

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

normal_data = np.random.normal(
    loc=50,
    scale=10,
    size=5000
)

normal_data[:10]
````

# 📈  Visualize Normal Data

````python
sns.histplot(
    normal_data,
    bins=40,
    kde=True
)

plt.title(
    "Approximately Symmetric Distribution"
)

plt.xlabel(
    "Value"
)

plt.ylabel(
    "Frequency"
)

plt.show()
````

### 👀 Observation

The distribution should look approximately symmetric.

Therefore, we expect:

```text
Skewness ≈ 0
```

---


# 📊  Calculate Skewness

Using Pandas:

````python
normal_series = pd.Series(
    normal_data
)

print(
    "Skewness:",
    normal_series.skew()
)
````

# 📊  Calculate Kurtosis

````python
print(
    "Excess Kurtosis:",
    normal_series.kurt()
)
````

# 📈  Create Right-Skewed Data

We can use an Exponential Distribution as a simple example of a strongly right-skewed distribution.


````python
right_skewed = np.random.exponential(
    scale=2,
    size=5000
)

right_skewed[:10]
````

# 📊  Visualize Right-Skewed Data

````python
sns.histplot(
    right_skewed,
    bins=40,
    kde=True
)

plt.title(
    "Right-Skewed Distribution"
)

plt.xlabel(
    "Value"
)

plt.ylabel(
    "Frequency"
)

plt.show()
````

### 👀 Observation

Most observations are concentrated near smaller values.

The distribution has a longer:

```text
Right Tail
```

Therefore:

```text
Positive Skewness
```

---

# 📐 Calculate Right-Skewness

````python
right_series = pd.Series(
    right_skewed
)

print(
    "Skewness:",
    right_series.skew()
)

print(
    "Excess Kurtosis:",
    right_series.kurt()
)
````

# 📉  Create Left-Skewed Data

We can create left-skewed data by reflecting a right-skewed dataset around a constant.

````python
left_skewed = (
    10 - right_skewed
)

left_skewed[:10]
````

# 📊  Visualize Left-Skewed Data

````python
sns.histplot(
    left_skewed,
    bins=40,
    kde=True
)

plt.title(
    "Left-Skewed Distribution"
)

plt.xlabel(
    "Value"
)

plt.ylabel(
    "Frequency"
)

plt.show()
````

### 👀 Observation

Most observations are concentrated toward larger values.

The longer tail extends toward the:

```text
Left side
```

Therefore:

```text
Negative Skewness
```

---

# 📐  Calculate Left-Skewness


````python
left_series = pd.Series(
    left_skewed
)

print(
    "Skewness:",
    left_series.skew()
)

print(
    "Excess Kurtosis:",
    left_series.kurt()
)
````

# 🧠  Compare the Three Distributions

Let's calculate their statistics together.


````python
summary = pd.DataFrame({
    "Distribution": [
        "Normal",
        "Right-Skewed",
        "Left-Skewed"
    ],
    "Skewness": [
        pd.Series(normal_data).skew(),
        pd.Series(right_skewed).skew(),
        pd.Series(left_skewed).skew()
    ],
    "Excess Kurtosis": [
        pd.Series(normal_data).kurt(),
        pd.Series(right_skewed).kurt(),
        pd.Series(left_skewed).kurt()
    ]
})

summary
````

# 📊  Understanding the Results

We expect approximately:

```text
Normal
Skewness ≈ 0
Excess Kurtosis ≈ 0
```

```text
Right-Skewed
Skewness > 0
```

```text
Left-Skewed
Skewness < 0
```

The exact values depend on the random sample.

---

# 🧠  SciPy Skewness

SciPy provides statistical functions through:

````python
from scipy import stats

# Let's calculate:

normal_skewness = stats.skew(
    normal_data
)

right_skewness = stats.skew(
    right_skewed
)

left_skewness = stats.skew(
    left_skewed
)

print(
    "Normal:",
    normal_skewness
)

print(
    "Right-Skewed:",
    right_skewness
)

print(
    "Left-Skewed:",
    left_skewness
)
````

# 📊 SciPy Kurtosis

````python
normal_kurtosis = stats.kurtosis(
    normal_data
)

right_kurtosis = stats.kurtosis(
    right_skewed
)

left_kurtosis = stats.kurtosis(
    left_skewed
)

print(
    "Normal:",
    normal_kurtosis
)

print(
    "Right-Skewed:",
    right_kurtosis
)

print(
    "Left-Skewed:",
    left_kurtosis
)
````

By default, SciPy's `kurtosis()` reports **Fisher's definition**, meaning:

```text
Normal Distribution → 0
```

This is excess kurtosis.

---


# 🧠  Pearson Kurtosis with SciPy

If we want Pearson Kurtosis:

```text
Normal = 3
```

we can use:


````python
normal_pearson_kurtosis = stats.kurtosis(
    normal_data,
    fisher=False
)

print(
    "Pearson Kurtosis:",
    normal_pearson_kurtosis
)
````

# 📊  Fisher vs Pearson Kurtosis

````python
fisher_kurtosis = stats.kurtosis(
    normal_data,
    fisher=True
)

pearson_kurtosis = stats.kurtosis(
    normal_data,
    fisher=False
)

print(
    "Excess Kurtosis:",
    fisher_kurtosis
)

print(
    "Pearson Kurtosis:",
    pearson_kurtosis
)
````

Remember:

```text
Pearson Kurtosis
=
Excess Kurtosis + 3
```

# ⚠️  Create Data with an Outlier

````python
data_without_outlier = np.array([
    10, 11, 12, 12, 13,
    13, 14, 14, 15, 15,
    16, 16, 17, 17, 18
])

data_with_outlier = np.append(
    data_without_outlier,
    100
)
````

# 🧠  Skewness and Feature Transformation

Suppose a feature is strongly right-skewed.

We may investigate transformations such as:

```text
Log Transformation
```

or:

```text
Power Transformation
```

The goal may be to make the distribution:

```text
More symmetric
```

and reduce the influence of extreme values.

---

# 📊 Log Transformation Example

Let's create right-skewed data.


````python
right_skewed = np.random.exponential(
    scale=2,
    size=5000
)

# Apply a log transformation:
log_transformed = np.log1p(
    right_skewed
)
````

Here:

```text
np.log1p(x)
```

calculates:

```text
log(1+x)
```

which is useful when values include zero.

---

# 📈  Compare Skewness Before and After


````python
before = stats.skew(
    right_skewed
)

after = stats.skew(
    log_transformed
)

print(
    "Skewness Before:",
    before
)

print(
    "Skewness After:",
    after
)
````

# 📊 Visualize Before Transformation

````python
sns.histplot(
    right_skewed,
    bins=40,
    kde=True
)

plt.title(
    "Before Log Transformation"
)

plt.xlabel(
    "Value"
)

plt.ylabel(
    "Frequency"
)

plt.show()
````

# 📊  Visualize After Transformation

````python
sns.histplot(
    log_transformed,
    bins=40,
    kde=True
)

plt.title(
    "After Log Transformation"
)

plt.xlabel(
    "Log-Transformed Value"
)

plt.ylabel(
    "Frequency"
)

plt.show()
````

### 👀 Observation

The log transformation compresses larger values more strongly.

This can reduce right skewness.

# ⚠️  Important Statistical Caution

Skewness and Kurtosis are:

```text
Descriptive Statistics
```

They do not prove that a distribution follows a particular probability distribution.

For example:

```text
Skewness ≈ 0
```

does not prove:

```text
Normal Distribution
```

Similarly:

```text
Kurtosis ≈ 0
```

does not prove Normality.

Visualizations and formal statistical tests may also be needed.

---

# 🧠  Skewness Does Not Equal Outlier Count

A dataset can be:

```text
Highly Skewed
```

without having one obvious isolated outlier.

Similarly:

```text
An Outlier
```

can increase skewness, but not every skewed dataset has an individual outlier.

Therefore:

```text
Skewness ≠ Outlier Detector
```

It is a distribution-shape measure.

---

# 📊  Kurtosis Does Not Equal Peak Height

Another important correction:

```text
Kurtosis ≠ Simply "Peakiness"
```

Modern statistical interpretation focuses strongly on:

```text
Tail Heaviness
```

and:

```text
Extreme Observation Tendency
```

This is a better mental model.

# 🧮  Formula Sheet

## 📐 Skewness

Population-style formula:

```text
γ₁
=
E[(X - μ)³]
────────────
σ³
```

### Interpretation

```text
γ₁ ≈ 0
→ Approximately Symmetric

γ₁ > 0
→ Right Skewed

γ₁ < 0
→ Left Skewed
```

---

# 📊  Kurtosis

Population-style Pearson Kurtosis:

```text
β₂
=
E[(X - μ)⁴]
────────────
σ⁴
```

For a Normal Distribution:

```text
Pearson Kurtosis = 3
```

---

# 📐  Excess Kurtosis

```text
Excess Kurtosis
=
Pearson Kurtosis - 3
```

Therefore:

```text
Normal → Excess Kurtosis = 0
```

### Interpretation

```text
Excess Kurtosis > 0
→ Heavier Tails

Excess Kurtosis ≈ 0
→ Normal-like Tail Behavior

Excess Kurtosis < 0
→ Lighter Tails
```

# 💎  Key Insights

- 📐 **Skewness measures asymmetry.**
- 📈 Positive Skewness means a longer right tail.
- 📉 Negative Skewness means a longer left tail.
- ⚖️ Symmetric distributions have Skewness near zero.
- 🧠 Skewness is based on the third central moment.
- 📊 **Kurtosis describes tail behavior and extreme-value tendency.**
- 🧮 Kurtosis is related to the fourth central moment.
- 📈 Positive Excess Kurtosis indicates heavier tails.
- 📉 Negative Excess Kurtosis indicates lighter tails.
- 🔔 A Normal Distribution has approximately:
  
```text
Skewness = 0
Excess Kurtosis = 0
```

- 📊 Pearson Kurtosis for a Normal Distribution is:

```text
3
```

- ⚠️ Skewness and Kurtosis do not prove Normality.
- 📦 Skewness and Kurtosis can be useful during EDA.
- 🔍 Extreme observations can strongly affect both measures.
- 🤖 Distribution analysis can help guide preprocessing and feature transformation decisions.

### ⭐ Remember

> **Skewness → Which side is the distribution stretched toward?**

> **Kurtosis → How heavy are the tails / how much extreme-value tendency is present?**
