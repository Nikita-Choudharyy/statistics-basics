# 🧠  What is a Random Variable?

Before understanding Probability Distributions, we need to understand:

# 🎯 Random Variable

A Random Variable is a variable whose value depends on the outcome of a random experiment.

For example, suppose we roll a die.

The possible outcomes are:

```text
1, 2, 3, 4, 5, 6
```

We can define a random variable:

```text
X = Number obtained from the die
```

Then:

```text
X ∈ {1, 2, 3, 4, 5, 6}
```

Here, `X` is a random variable.

 ---

# 🎲 Simple Random Variable Example

Suppose we toss a coin.

The possible outcomes are:

```text
Heads
Tails
```

We can define:

```text
X = Number of Heads
```

For one toss:

```text
X = 0 or 1
```

So:

```text
X ∈ {0, 1}
```

This is a random variable.

 ---


# 🧠  Why Do We Need Random Variables?

For example:

```text

```

```text

```



Random variables allow us to represent these outcomes mathematically.

# 📊 What is a Probability Distribution?

A Probability Distribution describes how probabilities are assigned to the possible values of a random variable.

For example, for a fair die:

| 🎲 Value | Probability |
|---:|---:|
| 1 | 1/6 |
| 2 | 1/6 |
| 3 | 1/6 |
| 4 | 1/6 |
| 5 | 1/6 |
| 6 | 1/6 |

The distribution tells us:

> Each possible outcome has a probability of `1/6`.

# 🎯  Probability Mass Function (PMF)

For discrete random variables, we commonly use a:

# 📊 Probability Mass Function

or:

```text
PMF
```

A PMF gives the probability of each possible value of a discrete random variable.

It is written as:

```text
P(X = x)
```

This means:

> Probability that random variable X takes the value x.

---

# 🎲  Example — PMF of a Fair Die

For a fair die:

```text
X ∈ {1, 2, 3, 4, 5, 6}
```

The PMF is:

| X | P(X = x) |
|---:|---:|
| 1 | 1/6 |
| 2 | 1/6 |
| 3 | 1/6 |
| 4 | 1/6 |
| 5 | 1/6 |
| 6 | 1/6 |

Therefore:

```text
P(X = 1) = 1/6
```

```text
P(X = 2) = 1/6
```

and so on.

---


# 🧮  Properties of a PMF

For a valid discrete probability distribution:

### Property 1

Every probability must be non-negative:

```text
P(X = x) ≥ 0
```

### Property 2

The total probability must equal 1:

```text
Σ P(X = x) = 1
```

Therefore:

```text
Total Probability = 1
```

---

# 📈  PMF Visualization

Let's visualize the probability distribution of a fair die.

````python
import numpy as np
import matplotlib.pyplot as plt

values = np.arange(1, 7)
probabilities = np.full(6, 1/6)

plt.bar(values, probabilities)

plt.title("PMF of a Fair Die")
plt.xlabel("Die Value")
plt.ylabel("Probability")

plt.xticks(values)

plt.show()
````

### 👀 Observation

Each value has the same probability:

```text
1/6 ≈ 0.1667
```

Therefore, the distribution is uniform across the six outcomes.

# 📊  Probability Density Function (PDF)

For continuous random variables, we commonly use a:

# 📈 Probability Density Function

or:

```text
PDF
```

A PDF describes the **density of probability across possible values**.

It is important to understand:

> For a continuous random variable, the probability of one exact value is generally zero.

For example:

```text
P(Height = exactly 170.000000... cm) ≈ 0
```

Instead, we calculate probabilities over intervals.

For example:

```text
P(170 cm < Height < 175 cm)
```

---


# 📏  Why is Exact Probability Zero?

A continuous variable can take infinitely many possible values.

Suppose height can be any value between:

```text
170 cm and 171 cm
```

There are infinitely many possible values:

```text
170.1
170.01
170.001
170.0001
...
```

Therefore, the probability assigned to one exact point is effectively zero.

But an interval can have positive probability.

For example:

```text
P(170 < Height < 171)
```

can be greater than zero.

---


# 🧠  PDF Does Not Directly Give Probability

This is a very important distinction.

For a continuous distribution:

```text
PDF value ≠ Probability
```

The probability of an interval is related to the **area under the PDF curve**.

For example:

```text
P(a < X < b)
```

corresponds to the area under the curve between `a` and `b`.

---

# 📊  Visual Intuition — PDF

Imagine a continuous distribution:

```text
Density
  │
  │             ╭──╮
  │           ╭─╯  ╰─╮
  │         ╭─╯      ╰─╮
  │       ╭─╯          ╰─╮
  │───────┴──────────────┴──────► X
          a              b
```

The area between:

```text
a
```

and:

```text
b
```

represents:

```text
P(a < X < b)
```

---


# 🧮  Properties of a PDF

For a valid continuous probability distribution:

### Property 1

The density cannot be negative:

```text
f(x) ≥ 0
```

### Property 2

The total area under the curve must equal:

```text
1
```

Mathematically:

```text
∫ f(x) dx = 1
```

across the entire possible range.

---


# ⚖️  PMF vs PDF

| 📊 PMF | 📈 PDF |
|---|---|
| Used for discrete variables | Used for continuous variables |
| Gives probability of exact values | Describes probability density |
| `P(X = x)` can be positive | `P(X = x)` is generally 0 |
| Values are countable | Values can vary continuously |
| Probability is assigned to individual outcomes | Probability is calculated over intervals |
| Example: Dice | Example: Height |

# 🐍  Creating a Discrete Dataset

Let's create a simple dataset representing the number of customers arriving at a store each hour.


````python
import numpy as np
import pandas as pd

customers_per_hour = np.array([
    4, 6, 5, 8, 7,
    4, 9, 6, 5, 7
])

df_discrete = pd.DataFrame({
    "Customers": customers_per_hour
})

df_discrete
````

The values are counts.

Therefore:

```text
Customers
```

is a discrete variable.

---

# 📊  Visualizing Discrete Data

````python
plt.hist(
    customers_per_hour,
    bins=np.arange(
        customers_per_hour.min(),
        customers_per_hour.max() + 2
    ) - 0.5,
    edgecolor="black"
)

plt.title("Distribution of Customers per Hour")
plt.xlabel("Number of Customers")
plt.ylabel("Frequency")

plt.show()
````

### 👀 Observation

The variable represents:

```text
Number of Customers
```

which is a count.

Therefore, it is discrete.

---

# 📊  Continuous Distribution with Seaborn

For a larger dataset, we can visualize a continuous distribution using Seaborn.


````python
import seaborn as sns

np.random.seed(42)

heights = np.random.normal(
    loc=170,
    scale=7,
    size=1000
)

sns.histplot(
    heights,
    kde=True
)

plt.title("Distribution of Heights")
plt.xlabel("Height")
plt.ylabel("Frequency")

plt.show()
````


### 👀 Observation

Height is a continuous measurement, and the values can vary across a range.

The KDE curve gives us a smooth approximation of the underlying distribution.

# 🧠 A Very Important Difference

For a discrete variable:

```text
P(X = x)
```

can be greater than zero.

For example:

```text
P(Die = 4) = 1/6
```

For a continuous variable:

```text

```

for an exact individual point in the usual continuous model.

Instead:

```text

```

can be greater than zero.

# 🎯 Example of Interval Probability

Suppose:

```text
X = Height
```

We may ask:

```text
P(160 < X < 170)
```

This is meaningful because we are asking for the probability that height lies within a range.

We are not asking for one exact measurement.

---

# 📊  Visualization of an Interval

````python
x = np.linspace(
    130,
    210,
    1000
)

mean = 170
std = 7

pdf = (
    1 / (std * np.sqrt(2 * np.pi))
    *
    np.exp(
        -0.5 * ((x - mean) / std) ** 2
    )
)

plt.plot(
    x,
    pdf
)

plt.axvline(
    160,
    linestyle="--"
)

plt.axvline(
    180,
    linestyle="--"
)

plt.title(
    "Continuous Distribution and Interval"
)

plt.xlabel("Height")
plt.ylabel("Density")

plt.show()
````


The area between:

```text
160
```

and:

```text
180
```

represents the probability that the continuous variable falls within that interval.

# ⚠️  Common Mistakes

### ❌ Mistake 1 — "All integers are discrete"

Not necessarily.

Always consider what the variable represents.

---

### ❌ Mistake 2 — "All decimals are continuous"

Not necessarily.

A count may be stored as a decimal because of data processing.

The meaning of the variable matters more.

---

### ❌ Mistake 3 — Thinking PDF value is probability

For continuous distributions:

```text
PDF value ≠ Probability
```

Probability corresponds to an area over an interval.
---

### ❌ Mistake 4 — Thinking continuous means infinite values everywhere

A continuous variable can have a defined range.

For example:

```text
Temperature ∈ [−50°C, 50°C]
```

It is still continuous because there can be infinitely many possible values within the range.

---

### ❌ Mistake 5 — Assuming every continuous variable is Normally Distributed

Continuous does not mean Normal.

A continuous variable may follow:

- Normal Distribution
- Uniform Distribution
- Exponential Distribution
- Gamma Distribution
- Log-Normal Distribution
- Other distributions

 ---

# 🧠  Quick Decision Rule

When you see a variable, ask:

### Question 1

> Is it being **counted**?

If yes:

```text
Usually Discrete
```

### Question 2

> Is it being **measured**?

If yes:

```text
Usually Continuous
```

### Question 3

> Can values exist between two possible values?

If yes:

```text
Likely Continuous
```

If the possible outcomes are separate and countable:

```text
Likely Discrete
```

# 💎 Key Insights

- 🎯 A **Random Variable** represents the numerical outcome of a random experiment.
- 📊 A **Probability Distribution** describes how probabilities are assigned across possible values.
- 🔢 **Discrete variables** are generally countable.
- 📏 **Continuous variables** are generally measurable.
- 🎲 Discrete distributions use a **PMF**.
- 📈 Continuous distributions use a **PDF**.

- 🧠 The meaning of a variable matters more than how it is stored in Python.
- 🤖 Understanding variable types is important for statistical modeling and Machine Learning.
- 📈 Different variables may require different probability distributions.

# 📌  Formula & Concept Summary

## 🔢 Discrete

For a discrete random variable:

```text
P(X = x)
```

represents the probability of a specific value.

The total probability is:

```text
Σ P(X = x) = 1
```



## 📏 Continuous

For a continuous random variable:

```text
P(X = x) = 0
```

for an exact point under the usual continuous model.

Probability over an interval is:

```text
P(a < X < b)
```

and corresponds to the area under the PDF between `a` and `b`.

The total area under the PDF is:

```text
∫ f(x) dx = 1
```
