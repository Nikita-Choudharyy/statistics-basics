# 🧠 What is a Poisson Distribution?

A Poisson Distribution describes the probability of observing a certain number of events within a fixed interval.

The interval could be:

### ⏱️ Time

```text
Number of calls per hour
```

### 📏 Distance

```text
Number of accidents per 100 km
```

### 📐 Area

```text
Number of defects per square meter
```

### 🌐 Space

```text
Number of website requests per second
```

The important idea is:

> **We are counting how many events occur within a specified interval.**

We usually write:


> X represents the number of events occurring in the chosen interval when the average event rate is 5.

# 📐  Poisson Probability Formula

The probability of observing exactly `k` events is:

```text
P(X = k)
=
(e^(-λ) λ^k) / k!
```

Where:

```text
X = Number of events

k = Specific number of events

λ = Average number of events

e = Euler's number

k! = Factorial of k
```

---

# 🧮  Understanding the Formula

The formula:

```text
P(X = k)
=
(e^(-λ) λ^k) / k!
```

has three important components:

```text
λ
↓
Average event rate
```

```text
k
↓
Number of events we are interested in
```

```text
k!
↓
Factorial adjustment
```

The formula gives the probability of observing exactly `k` events.

---

# 📞  Example — Exactly 3 Calls

Suppose a support center receives:

```text
λ = 5 calls/hour
```

We want:

```text
P(X = 3)
```

Using the formula:

```text
P(X = 3)
=
(e^-5 × 5³) / 3!
```

Since:

```text
3! = 6
```

we can calculate this probability using Python.

 ---

# 🐍 Calculate Poisson Probability Manually


````python
import math

lambda_value = 5
k = 3

probability = (
    math.exp(-lambda_value)
    * (lambda_value ** k)
    / math.factorial(k)
)

print(
    "P(X = 3):",
    probability
)
````

# 📊  Understanding the Result

The result represents:

```text
P(X = 3)
```

In simple words:

> The probability of receiving exactly 3 calls in one hour when the average number of calls per hour is 5.

# 🧠  Poisson Distribution Conditions

The Poisson model is based on assumptions about how events occur.

Important ideas include:

### 1️⃣ Fixed Interval

We define a specific interval.

For example:

```text
1 hour
```

or:

```text
1 kilometer
```

or:


```text
1 square meter
```



### 2️⃣ Count Events

We count how many times an event occurs.

For example:

```text
Number of calls
```

### 3️⃣ Events Occur Independently

The occurrence of one event should not directly affect another event under the model.

### 4️⃣ Stable Average Rate

The average event rate should be reasonably stable over the interval being modeled.

For example:

```text
Average = 5 calls/hour
```

should be a reasonable representation of the process during the period being analyzed.


### 5️⃣ Events Occur One at a Time

The standard Poisson process assumes events occur individually rather than as simultaneous grouped events.

---

Think:

```text
📦 Fixed Interval
      ↓
🔢 Count Events
      ↓
🔗 Independent Occurrence
      ↓
📊 Stable Average Rate
      ↓
🎯 Poisson Model
```

```text
P(X = k)
```

# 🐍  Poisson Distribution with SciPy

SciPy provides the Poisson distribution through:


````python
from scipy.stats import poisson

# Let's create a distribution:

lambda_value = 5

distribution = poisson(
    mu=lambda_value
)
````

Here:

```text
mu = λ
```

---


# 🎯  Probability of Exactly k Events

Suppose:

```text
λ = 5
```

and we want:

```text
P(X = 3)
```

Use:


````python
probability = distribution.pmf(3)

print(
    "P(X = 3):",
    probability
)
````

# 📊  Generate Poisson PMF

Let's calculate probabilities for several possible values.


````python
import numpy as np

x = np.arange(
    0,
    16
)

probabilities = distribution.pmf(x)

print(probabilities)
````

Here:

```text
x
```

represents the number of events.

# 🔬  Verify Using SciPy

````python
distribution = poisson(
    mu=5
)

print(
    "Mean:",
    distribution.mean()
)

print(
    "Variance:",
    distribution.var()
)

print(
    "Standard Deviation:",
    distribution.std()
)
````

# 🧠  Poisson vs Binomial

Poisson and Binomial are both discrete distributions, but they model different situations.

### 🎯 Binomial

We know:

```text
Number of trials = n
```

and:

```text
Probability of success = p
```

Then we count:

```text
Number of successes
```

---

### 📦 Poisson

We know:

```text
Average event rate = λ
```

over a specified interval.

Then we count:

```text
Number of events
```

---


# 📊  Comparison Table

| 🎯 Binomial | 📦 Poisson |
|---|---|
| Fixed number of trials | Fixed interval |
| Two outcomes per trial | Count events |
| Parameters `n, p` | Parameter `λ` |
| Counts successes | Counts events |
| Example: Successful customers among 100 customers | Example: Calls per hour |
| Mean = `np` | Mean = `λ` |
| Variance = `np(1-p)` | Variance = `λ` |

---

# 🔗  Relationship Between Binomial and Poisson

The Poisson Distribution can be viewed as a limiting case of the Binomial Distribution under certain conditions.

When:

```text
n → very large
```

and:

```text
p → very small
```

while:

```text
np = λ
```

remains approximately constant:

```text
Binomial(n,p)
        ↓
Large n
Small p
        ↓
Poisson(λ)
```

This is called the:

### ⭐ Poisson Approximation to the Binomial

---

# 🧠  Example of Poisson Approximation

Suppose:

```text
n = 1000
p = 0.002
```

Then:

```text
np = 1000 × 0.002
```

```text
= 2
```

So we can use:

```text
λ = 2
```

as a Poisson approximation under suitable conditions.

---

# 📊  Why is This Useful?

Sometimes calculating a Binomial probability with:

```text
Very large n
```

can be inconvenient.

If:

```text
n is large
p is small
np is moderate
```

the Poisson approximation can simplify calculations.

# 🧮  Poisson CDF

The PMF gives:

```text
P(X = k)
```

The CDF gives:

```text
P(X ≤ k)
```

For example:

````python
lambda_value = 5

probability = poisson.cdf(
    4,
    mu=lambda_value
)

print(
    "P(X <= 4):",
    probability
)
````

# 🎯  At Least k Events

Suppose:

```text
λ = 5
```

We want:

```text
P(X ≥ 6)
```

We can use:

```text
P(X ≥ 6)
=
1 - P(X ≤ 5)
```

In Python:


````python
probability = (
    1 -
    poisson.cdf(
        5,
        mu=5
    )
)

print(
    "P(X >= 6):",
    probability
)
````

# 🧠  Exactly vs At Most vs At Least

| Wording | Mathematical Form | Python |
|---|---|---|
| Exactly 5 | `P(X = 5)` | `poisson.pmf(5, μ)` |
| At most 5 | `P(X ≤ 5)` | `poisson.cdf(5, μ)` |
| At least 5 | `P(X ≥ 5)` | `1 - poisson.cdf(4, μ)` |
| More than 5 | `P(X > 5)` | `1 - poisson.cdf(5, μ)` |
| Less than 5 | `P(X < 5)` | `poisson.cdf(4, μ)` |

# 🎲  Simulating Poisson Events

NumPy allows us to simulate Poisson-distributed data.


````python
np.random.seed(42)

events = np.random.poisson(
    lam=5,
    size=10000
)
````

Here:

```text
lam = Average number of events
size = Number of simulated intervals
```

Each value represents:

```text
Number of events in one interval
```

---

# 📊  Visualize the Simulation

````python
plt.hist(
    events,
    bins=np.arange(
        events.min() - 0.5,
        events.max() + 1.5,
        1
    ),
    edgecolor="black"
)

plt.title(
    "Simulated Poisson Distribution"
)

plt.xlabel(
    "Number of Events"
)

plt.ylabel(
    "Frequency"
)

plt.show()
````

# 🧮 Experimental Mean

Let's calculate the mean of the simulated data.

````python
print(
    "Experimental Mean:",
    np.mean(events)
)
````

# 📊  Experimental Variance

````python
print(
    "Experimental Variance:",
    np.var(events)
)
````

# 🧠  Theoretical vs Experimental

For:

```text
λ = 5
```

the theoretical values are:

```text
Mean = 5
Variance = 5
```

The simulated values may be slightly different because of random variation.

As the number of simulations increases, the experimental results generally become closer to the theoretical values.

---

# 📊  Comparing Theoretical and Experimental PMF


````python
lambda_value = 5

x = np.arange(
    0,
    16
)

theoretical = poisson.pmf(
    x,
    mu=lambda_value
)

experimental = np.array([
    np.mean(events == value)
    for value in x
])
````

# 📈  Plot Comparison

````python
plt.bar(
    x,
    theoretical,
    alpha=0.6,
    label="Theoretical"
)

plt.plot(
    x,
    experimental,
    marker="o",
    label="Experimental"
)

plt.title(
    "Theoretical vs Experimental Poisson Distribution"
)

plt.xlabel(
    "Number of Events"
)

plt.ylabel(
    "Probability"
)

plt.legend()

plt.show()
````


### 👀 Observation

The experimental probabilities should approximately follow the theoretical Poisson PMF.

# ⚠️ Common Mistakes

### ❌ Mistake 1 — Using Poisson for Continuous Measurements

Poisson is for:

```text
Counts
```

not measurements such as:

```text
Height
Weight
Temperature
```

---

### ❌ Mistake 2 — Thinking λ is the Exact Number of Events

If:

```text
λ = 5
```

it means:

```text
Average = 5
```

not:

```text
Exactly 5 every time
```

---

### ❌ Mistake 3 — Forgetting the Interval

Saying:

```text
λ = 5
```

is incomplete without understanding:

```text
5 events per what?
```

It could be:

```text
5/hour
5/minute
5/km
5/product
```

---

### ❌ Mistake 4 — Confusing λ with Probability

Lambda is an:

```text
Average Event Rate
```

It is not restricted to:

```text
0 ≤ λ ≤ 1
```

For example:

```text
λ = 100
```

is completely valid.

---

### ❌ Mistake 5 — Assuming Mean = Variance Proves Poisson

Although:

```text
Mean = Variance
```

is a key Poisson property, matching these values alone does not prove that a dataset follows a Poisson distribution.

---

### ❌ Mistake 6 — Ignoring Changing Event Rates

If the event rate changes dramatically across the interval, a simple Poisson model may not be appropriate.

For example:

```text
Daytime → 20 calls/hour
Night → 2 calls/hour
```

A single constant-rate Poisson model may not describe the entire period well.

# 💎 Key Insights

- 📦 Poisson Distribution models the **number of events in a fixed interval**.
- 🔢 It is a **discrete probability distribution**.
- λ represents the **average number of events in the interval**.
- 📊 The Poisson PMF gives the probability of exactly `k` events.
- 🧮 The main formula is:

```text
P(X = k)
=
(e^(-λ) λ^k) / k!
```

- 🎯 For Poisson:

```text
Mean = λ
```

- 📊 Also:

```text
Variance = λ
```

- 📏 Therefore:
```text
SD = √λ
```

- 🔗 Poisson can approximate Binomial when the number of trials is large and the success probability is small, with `np` approximately constant.
- 📈 As λ becomes larger, the distribution becomes more symmetric and can resemble a Normal Distribution.
- 🤖 Poisson models are especially relevant for count data and Poisson Regression.
- ⚠️ The assumptions behind the Poisson model matter.

### ⭐ Remember

> **Binomial → Count successes in a fixed number of trials.**

> **Poisson → Count events in a fixed interval.**
