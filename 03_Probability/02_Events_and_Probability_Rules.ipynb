```text

Event        → Selected outcomes
```

# 🔗  Union of Two Events

The **Union** of two events means:

> Event A **OR** Event B occurs.

The union is represented by:

```text
A ∪ B
```

We read this as:

> "A union B"

The union contains all outcomes that belong to:

```text
A
OR
B
OR
both A and B
```

---

# 🎲  Example of Union

Roll a fair six-sided die.

Sample space:

```text
S = {1, 2, 3, 4, 5, 6}
```

Let:

```text
A = {2, 4, 6}
```

Event A:

> Getting an even number.

And:

```text
B = {4, 5, 6}
```

Event B:

> Getting a number greater than 3.

Then:

```text
A ∪ B
```

contains:

```text
{2, 4, 5, 6}
```

Notice that `4` and `6` belong to both events, but they are counted only once in the union.

# ✖️ Intersection of Two Events

The **Intersection** of two events means:

> Event A **AND** Event B occur together.

It is represented by:

```text
A ∩ B
```

We read this as:

> "A intersection B"

The intersection contains only outcomes that belong to **both A and B**.

 ---


# 🎲 Example of Intersection

Again, roll a fair die.

Let:

```text
A = {2, 4, 6}
```

A means:

> Getting an even number.

And:

```text
B = {4, 5, 6}
```

B means:

> Getting a number greater than 3.

The values common to both are:

```text
{4, 6}
```

Therefore:

```text
A ∩ B = {4, 6}
```

---


# 🧠 Union vs Intersection

This is one of the most important distinctions.

| 🔗 Union | ✖️ Intersection |
|---|---|
| A OR B | A AND B |
| `A ∪ B` | `A ∩ B` |
| Includes A | Includes only common outcomes |
| Includes B | Includes only overlapping outcomes |
| Includes overlap | Only overlap |

### 🔑 Remember

```text
OR  → Union
AND → Intersection
```

---

# 🔄  Complement of an Event

The **Complement** of an event means that the event does **not** occur.

It is commonly represented as:

```text
Aᶜ
```

or:

```text
A'
```

or:

```text
Not A
```

All represent the idea:

> A does not happen.

---


# 🎲  Complement Example

Roll a die.

Sample space:

```text
S = {1, 2, 3, 4, 5, 6}
```

Suppose:

```text
A = {2, 4, 6}
```

A means:

> Getting an even number.

The complement is:

```text
Aᶜ = {1, 3, 5}
```

which means:

> Getting an odd number.

Therefore:

```text
A + Aᶜ = Entire Sample Space
```

In probability:

```text
P(Aᶜ) = 1 − P(A)
```

---

# 📐 Complement Rule

The probability of the complement of an event is:

```text
P(Aᶜ) = 1 − P(A)
```

Where:

- `P(A)` = Probability that A occurs
- `P(Aᶜ)` = Probability that A does not occur

This works because:

```text
Event A + Not A = Certain Event
```

Therefore:

```text
P(A) + P(Aᶜ) = 1
```

 ---

# 🎲 Complement Example

Suppose:

```text
P(Rain) = 0.30
```

Then:

```text
P(No Rain) = 1 − 0.30
           = 0.70
```

Therefore:

```text
P(No Rain) = 70%
```

### 💡 Why is this useful?

Sometimes calculating an event directly is difficult, while calculating its complement is easier.

For example:

> Probability of at least one success

can often be calculated using:

```text
1 − Probability of no success
```

This idea becomes very useful later in probability problems.

---

# ⚖️ Mutually Exclusive Events

Two events are **Mutually Exclusive** if they cannot happen at the same time.

In other words:

```text
A ∩ B = ∅
```

where:

```text
∅ = Empty Set
```

Therefore:

```text
P(A ∩ B) = 0
```

---

# 🪙 Example — Heads and Tails

For one coin toss:

```text
A = Heads
B = Tails
```

A single coin toss cannot result in both Heads and Tails at the same time.

Therefore:

```text
A ∩ B = ∅
```

and:

```text
P(A ∩ B) = 0
```

So Heads and Tails are **mutually exclusive events** for a single toss.

---


# 🎲 Another Example — Die Roll

Suppose:

```text
A = Rolling a 2
B = Rolling a 5
```

A single die roll cannot be both 2 and 5.

Therefore:

```text
A ∩ B = ∅
```

These events are mutually exclusive.

---

# ⚠️  Mutually Exclusive Does NOT Mean Independent

This is an important distinction.

If two events are mutually exclusive and both have positive probability, they **cannot be independent**.

Why?

Because if A occurs, B cannot occur.

For mutually exclusive events:

```text
P(A ∩ B) = 0
```

But for independent events:

```text
P(A ∩ B) = P(A)P(B)
```

If both probabilities are positive, then:

```text
P(A)P(B) > 0
```

So the two conditions cannot both hold.

### 🔑 Remember

> **Mutually Exclusive → Cannot happen together**

> **Independent → One event does not affect the probability of the other**

We will study independence more deeply in the next probability notebooks.

---



# ➕  Addition Rule of Probability

The **Addition Rule** helps us calculate the probability of:

```text
A OR B
```

For two general events:

```text
P(A ∪ B)
=
P(A) + P(B) − P(A ∩ B)
```

The intersection is subtracted because the overlapping outcomes would otherwise be counted twice.

 ---

# 🧮 Why Do We Subtract the Intersection?

Suppose:

```text
P(A) = 0.50
P(B) = 0.40
P(A ∩ B) = 0.20
```

If we simply add:

```text
0.50 + 0.40 = 0.90
```

the overlapping part has been counted twice.

Therefore we subtract it once:

```text
P(A ∪ B)
= 0.50 + 0.40 − 0.20
= 0.70
```

Therefore:

```text
P(A ∪ B) = 0.70
```

---

# 🎲 Addition Rule Example

Consider a fair six-sided die.

Let:

```text
A = Getting an even number
```

Therefore:

```text
A = {2, 4, 6}
```

And:

```text
B = Getting a number greater than 3
```

Therefore:

```text
B = {4, 5, 6}
```

We have:

```text
P(A) = 3/6
P(B) = 3/6
```

The intersection is:

```text
A ∩ B = {4, 6}
```

Therefore:

```text
P(A ∩ B) = 2/6
```

Using the Addition Rule:

```text
P(A ∪ B)
= P(A) + P(B) − P(A ∩ B)
```

```text
= 3/6 + 3/6 − 2/6
```

```text
= 4/6
```

Therefore:

```text
P(A ∪ B) = 2/3
```

---

# ⚖️ Addition Rule for Mutually Exclusive Events

If A and B are mutually exclusive:

```text
P(A ∩ B) = 0
```

Therefore the general Addition Rule becomes:

```text
P(A ∪ B) = P(A) + P(B)
```

### Example

For a die:

```text
A = Rolling 1
B = Rolling 6
```

These events are mutually exclusive.

Therefore:

```text
P(A) = 1/6
P(B) = 1/6
```

So:

```text
P(A ∪ B)
= 1/6 + 1/6
= 2/6
= 1/3
```

---


# 🧠 General vs Mutually Exclusive Addition Rule

| 🔗 General Events | ⚖️ Mutually Exclusive Events |
|---|---|
| `P(A ∪ B) = P(A) + P(B) − P(A ∩ B)` | `P(A ∪ B) = P(A) + P(B)` |
| Events can overlap | Events cannot overlap |
| Intersection may be > 0 | Intersection = 0 |

### 🔑 Important

Always check whether events overlap before using the simplified rule.

---


# ✖️  Multiplication Rule of Probability

The Multiplication Rule is related to:

```text
A AND B
```

For general events:

```text
P(A ∩ B) = P(A)P(B|A)
```

This means:

> Probability that A and B both occur equals the probability of A multiplied by the probability of B given A.

The conditional probability term will be studied in detail in the next notebook.

---

# 🤖 Independent Events — Special Multiplication Rule

If A and B are **independent**, then:

```text
P(B|A) = P(B)
```

Therefore:

```text
P(A ∩ B) = P(A)P(B)
```

This is the multiplication rule for independent events.

 ---

# 🪙 Example — Two Coin Tosses

Suppose we toss a fair coin twice.

Let:

```text
A = Heads on first toss
B = Heads on second toss
```

The first toss does not affect the second toss.

Therefore, the events are independent.

We know:

```text
P(A) = 1/2
P(B) = 1/2
```

Therefore:

```text
P(A ∩ B)
= P(A)P(B)
```

```text
= 1/2 × 1/2
```

```text
= 1/4
```

Therefore:

```text
P(Two Heads) = 25%
```

---


# 🎲 Example — Two Dice

Suppose we roll two fair dice.

Let:

```text
A = First die shows 6
B = Second die shows 6
```

The outcome of one die does not affect the other.

Therefore:

```text
P(A) = 1/6
P(B) = 1/6
```

Using the independent multiplication rule:

```text
P(A ∩ B)
= 1/6 × 1/6
= 1/36
```

Therefore:

```text
P(Both dice show 6) = 1/36
```

---

# 🔄 Complement and Union

The complement rule and union rule are closely related.

For any event:

```text
P(A) + P(Aᶜ) = 1
```

Therefore:

```text
P(Aᶜ) = 1 − P(A)
```

This simple rule can solve many probability problems.

# 🧮 Calculating Probability from Sets

Suppose the sample space is:

````python
S = {1, 2, 3, 4, 5, 6}

A = {2, 4, 6}
B = {4, 5, 6}

p_A = len(A) / len(S)
p_B = len(B) / len(S)

print("P(A):", p_A)
print("P(B):", p_B)
````

# 🔗 Probability of Union Using Python

````python
union = A | B

p_union = len(union) / len(S)

print("A ∪ B:", union)
print("P(A ∪ B):", p_union)
````

# 🤝 Probability of Intersection Using Python

````python
intersection = A & B

p_intersection = len(intersection) / len(S)

print("A ∩ B:", intersection)
print("P(A ∩ B):", p_intersection)
````

# 🧪  Verify the Addition Rule with Python

Let's verify:

```text
P(A ∪ B) = P(A) + P(B) − P(A ∩ B)
```


````python
p_A = len(A) / len(S)
p_B = len(B) / len(S)
p_intersection = len(A & B) / len(S)

p_union_formula = (
    p_A +
    p_B -
    p_intersection
)

p_union_actual = len(A | B) / len(S)

print("Using Formula:", p_union_formula)
print("Actual:", p_union_actual)
````

Both values should be equal.

This demonstrates the Addition Rule computationally.

---

# 🎲 Simulating the Addition Rule

Let's simulate many dice rolls.

````python
import numpy as np

rolls = np.random.randint(1, 7, size=100000)

event_A = rolls % 2 == 0
event_B = rolls > 3

p_A = np.mean(event_A)
p_B = np.mean(event_B)

p_intersection = np.mean(
    event_A & event_B
)

p_union = np.mean(
    event_A | event_B
)

print("P(A):", p_A)
print("P(B):", p_B)
print("P(A ∩ B):", p_intersection)
print("P(A ∪ B):", p_union)
````

# 📊 Comparing Formula and Simulation

Let's calculate the union using the Addition Rule.

````python
p_union_formula = (
    p_A +
    p_B -
    p_intersection
)

print("Union from Simulation:", p_union)
print("Union from Formula:", p_union_formula)
````

The two values should be very close.

They may not be exactly equal because the simulation uses a finite number of random trials.

### 🔑 Important

> **Simulation approximates probability, while mathematical formulas provide the theoretical result under their assumptions.**

# 🧠 Important Distinction — OR vs AND

Whenever you see a probability question, pay close attention to the wording.

### OR

```text

```

Usually means:

```text

```

and uses the Addition Rule.

### AND

```text

```

Usually means:


### Example



# 🧠 Common Mistakes

### ❌ Mistake 1

Adding probabilities without considering overlap.

Wrong:

```text
P(A ∪ B) = P(A) + P(B)
```

for general overlapping events.

Correct:

```text
P(A ∪ B)
=
P(A) + P(B) − P(A ∩ B)
```

---

### ❌ Mistake 2

Confusing Union and Intersection.

Remember:

```text
OR  → Union
AND → Intersection
```

---

### ❌ Mistake 3

Thinking mutually exclusive means independent.

They are different concepts.

```text
Mutually Exclusive
→ Cannot happen together

Independent
→ One does not affect the other
```

---

### ❌ Mistake 4

Automatically treating every event as independent.

Independence must be justified by the experiment or assumptions.

---

### ❌ Mistake 5

Forgetting the overlap.

When two events overlap, the shared region gets counted twice if we simply add probabilities.

That is why the intersection is subtracted.

---

# 📌  Summary

In this notebook, we learned how events interact in Probability.

### 🔗 Union

```text
A ∪ B
```

means:

```text
A OR B
```

### ✖️ Intersection

```text
A ∩ B
```

means:

```text
A AND B
```

### 🔄 Complement

```text
Aᶜ
```

means:

```text
NOT A
```

### ➕ Addition Rule

```text
P(A ∪ B) = P(A) + P(B) − P(A ∩ B)
```

### ⚖️ Mutually Exclusive Events

If:

```text
A ∩ B = ∅
```

then:

```text
P(A ∪ B) = P(A) + P(B)
```

### ✖️ Independent Events

If A and B are independent:

```text
P(A ∩ B) = P(A)P(B)
```

> **Probability Rules help us combine and reason about multiple uncertain events.**

The most important mental model is:

```text
OR
 ↓
Union
 ↓
A ∪ B

AND
 ↓
Intersection
 ↓
A ∩ B

NOT
 ↓
Complement
 ↓
Aᶜ
```

And remember:

```text
General Union:
P(A ∪ B) = P(A) + P(B) − P(A ∩ B)
```

These concepts are the foundation for more advanced probability.
