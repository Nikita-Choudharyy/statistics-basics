# 🧠 The Core Idea of Bayes' Theorem

Suppose we want to know:

```text
P(A | B)
```

This means:

> Probability of A given that B has occurred.

Sometimes it is difficult to calculate `P(A | B)` directly.

But perhaps we already know:

```text
P(B | A)
P(A)
P(B)
```

Bayes' Theorem allows us to calculate:

```text
P(A | B)
```

using this information.

The main idea is:

```text
Initial Probability
        ↓
     New Evidence
        ↓
 Updated Probability
```

---


# 📐  Bayes' Theorem Formula

The formula is:


Where:

```text
P(A | B)
```

= Probability of A given B

```text
P(B | A)
```

= Probability of B given A

```text
P(A)
```

= Probability of A before observing B

```text
P(B)
```

= Probability of observing B

---


# 🧩 Understanding Each Part

Bayes' Theorem can be understood through four important terms:

```text
Prior
Likelihood
Evidence
Posterior
```

Let's understand each one.

---

# 🟡  Prior Probability

The **Prior Probability** represents our initial probability before considering new evidence.

It is:

```text
P(A)
```

For example:

Suppose a disease affects:

```text
1% of a population
```

Then:

```text
P(Disease) = 0.01
```

This is the **Prior Probability**.

It tells us:

> Before seeing any test result, what is the probability that a randomly selected person has the disease?

---

# 🔵 Likelihood

The **Likelihood** represents the probability of observing the evidence if the event is actually true.

It is:

```text
P(B | A)
```

For example:

```text
A = Person has the disease
B = Test is positive
```

Then:

```text
P(B | A)
```

means:

> Probability that the test is positive given that the person actually has the disease.

This is related to the test's ability to correctly detect the disease.

---

# 🟣 Evidence

The **Evidence** represents the overall probability of observing the evidence.

It is:

```text
P(B)
```

For example:

```text
P(Positive Test)
```

This includes positive test results from:

- People who have the disease
- People who do not have the disease

Therefore, it is the total probability of seeing a positive test.

---


# 🟢  Posterior Probability

The **Posterior Probability** is the updated probability after observing new evidence.

It is:

```text
P(A | B)
```

For example:

```text
P(Disease | Positive Test)
```

means:

> Probability that the person has the disease given that their test is positive.

This is usually the probability we are most interested in.

---

# 🧠 Bayes' Theorem Concept Map

```text
                 New Evidence
                      │
                      ▼
             ┌─────────────────┐
             │   Likelihood     │
             │     P(B | A)     │
             └────────┬────────┘
                      │
                      ▼
Prior ───────────► Bayes' Theorem
P(A)                   │
                       ▼
                Posterior
                P(A | B)
```

Another way to remember it:

```text
Prior
  +
Evidence
  ↓
Updated Belief
```

# 🔄 Why Do We Need Bayes' Theorem?

Suppose a medical test is positive.

A common mistake is to think:

> "The test is 95% accurate, so there is a 95% chance that the person has the disease."

That conclusion may be wrong.

Why?

Because we also need to know:

```text
How common is the disease?
```

This is where the **Prior Probability** matters.

Bayes' Theorem combines:

```text
Disease Prevalence
+
Test Characteristics
+
Observed Evidence
```

to calculate:

```text
Updated Probability
```

---


# 🏥 Medical Diagnosis Example

Let's consider a simplified medical testing problem.

Suppose:

```text
Disease prevalence = 1%
```

Therefore:

```text
P(Disease) = 0.01
```

Suppose the test has:

```text
Sensitivity = 95%
```

This means:

```text
P(Positive | Disease) = 0.95
```

Suppose the false positive rate is:

```text
5%
```

Therefore:

```text
P(Positive | No Disease) = 0.05
```

Now a person receives a positive test.

We want to calculate:

```text
P(Disease | Positive)
```

---


# 📐  Define the Events

Let:

```text
D = Person has the disease
```

and:

```text
P = Test is positive
```

We know:

```text
P(D) = 0.01
```

```text
P(P | D) = 0.95
```

```text
P(P | Dᶜ) = 0.05
```

We want:

```text
P(D | P)
```

---


# 🧠  Calculating the Evidence

To use Bayes' Theorem, we need:

```text
P(P)
```

The positive test can occur in two situations:

```text
Person has disease
OR
Person does not have disease
```

Therefore:

```text
P(P) = P(P | D)P(D) + P(P | Dᶜ)P(Dᶜ)
```

We know:

```text
P(D) = 0.01
```

Therefore:

```text
P(Dᶜ) = 1 - 0.01
       = 0.99
```

Now:

```text
P(P) = (0.95)(0.01) + (0.05)(0.99)
```

```text
= 0.0095 + 0.0495
```

```text
= 0.059
```

So:

```text
P(Positive) = 0.059
```

---

# 🧮 Applying Bayes' Theorem

Now:

```text
P(D | P) = P(P | D)P(D) / P(P)
```

Substitute the values:

```text
P(D | P) = (0.95 × 0.01) / 0.059
```

```text
≈ 0.161
```

Therefore:

```text
P(Disease | Positive)
≈ 16.1%
```

### 😮 Important Observation

Even though the test has:

```text
95% sensitivity
```

a positive test does **not** mean there is a 95% probability of disease.

The actual probability is approximately:

```text
16.1%
```

because the disease is relatively rare in the population.

---


# 🧠  Why Did This Happen?

The important factor is:

```text
Prior Probability
```

The disease affects only:

```text
1%
```

of the population.

Therefore, even a test with good performance can produce many positive results among people who do not have the disease when the disease is rare.

This is one of the most important practical lessons of Bayes' Theorem.

---

# 🧮  General Bayes' Theorem

For two events:

```text
P(A | B)
=
P(B | A)P(A)
────────────────
P(B)
```

But when the evidence B can occur through multiple mutually exclusive cases, we can expand the denominator.

For example:

```text
P(B)
=
P(B | A)P(A)
+
P(B | Aᶜ)P(Aᶜ)
```

Therefore:

```text
P(A | B)
=
P(B | A)P(A)
────────────────────────────
P(B | A)P(A) + P(B | Aᶜ)P(Aᶜ)
```

This form is very useful for binary classification problems.

# 🐍  Bayes' Theorem Using Python

Let's implement the medical example.

````python
prior = 0.01
sensitivity = 0.95
false_positive_rate = 0.05

# Calculate the probability of a positive test:

prob_positive = (
    sensitivity * prior
    +
    false_positive_rate * (1 - prior)
)

print("P(Positive):", prob_positive)
````

# 🧮  Calculate Posterior Probability

````python
posterior = (
    sensitivity * prior
) / prob_positive

print(
    "P(Disease | Positive):",
    posterior
)

# Convert it to percentage:

print(
    "Posterior Probability:",
    posterior * 100,
    "%"
)
````

# 📊 Visualizing Prior and Posterior

````python
import matplotlib.pyplot as plt

probabilities = [
    prior,
    posterior
]

labels = [
    "Prior",
    "Posterior"
]

plt.bar(labels, probabilities)

plt.title(
    "Bayesian Probability Update"
)

plt.ylabel("Probability")

plt.show()
````

### 👀 Observation

The probability changes after observing the evidence.

```text
Prior
1%

        ↓ Positive Test

Posterior
≈16.1%
```

This is the central idea of Bayesian updating.

---

# 🧮  Bayes' Theorem as a Python Function

We can create a reusable function.

````python
def bayes_theorem(
    prior,
    likelihood,
    evidence
):
    posterior = (
        likelihood * prior
    ) / evidence

    return posterior
````

````python
posterior = bayes_theorem(
    prior=0.01,
    likelihood=0.95,
    evidence=0.059
)

print(posterior)
````

# 🔍  A More Complete Bayes Function

We can calculate the evidence inside the function.

````python
def calculate_bayes_probability(
    prior,
    likelihood,
    false_positive_rate
):
    probability_evidence = (
        likelihood * prior
        +
        false_positive_rate * (1 - prior)
    )

    posterior = (
        likelihood * prior
    ) / probability_evidence

    return posterior
````

````python
posterior = calculate_bayes_probability(
    prior=0.01,
    likelihood=0.95,
    false_positive_rate=0.05
)

print(
    "Posterior:",
    posterior
)
````

# 📊 How Prior Probability Affects Posterior

Let's see what happens when disease prevalence changes.

````python
priors = [
    0.001,
    0.01,
    0.05,
    0.10,
    0.20
]

posteriors = []

for prior_value in priors:
    posterior_value = calculate_bayes_probability(
        prior=prior_value,
        likelihood=0.95,
        false_positive_rate=0.05
    )

    posteriors.append(
        posterior_value
    )

posteriors
````

# 📈  Visualizing Prior vs Posterior

````python
plt.plot(
    priors,
    posteriors,
    marker="o"
)

plt.title(
    "Effect of Prior Probability on Posterior"
)

plt.xlabel("Prior Probability")
plt.ylabel("Posterior Probability")

plt.show()
````

### 💡 Observation

The posterior probability depends strongly on the prior probability.

This demonstrates why:

> **Base rates matter.**

# 🔗  Bayes' Theorem and Conditional Probability

Bayes' Theorem comes directly from the relationship between two conditional probabilities.

We know:

```text
P(A ∩ B)
=
P(A | B)P(B)
```

and:

```text
P(A ∩ B)
=
P(B | A)P(A)
```

Since both represent the same intersection:

```text
P(A | B)P(B)
=
P(B | A)P(A)
```

Rearranging:

```text
P(A | B)
=
P(B | A)P(A)
────────────
P(B)
```

This is Bayes' Theorem.

### 💡 Important Insight

> **Bayes' Theorem is not a completely separate idea. It is derived from Conditional Probability.**

---


# 🧠  Derivation of Bayes' Theorem

Start with:

```text
P(A | B)
=
P(A ∩ B)
────────
P(B)
```

We also know:

```text
P(B | A)
=
P(A ∩ B)
────────
P(A)
```

Therefore:

```text
P(A ∩ B)
=
P(B | A)P(A)
```

Substitute this into the first equation:

```text
P(A | B)
=
P(B | A)P(A)
────────────
P(B)
```

Therefore:

# ⭐ Bayes' Theorem

```text
P(A | B)
=
P(B | A)P(A)
────────────
P(B)
```

# ⚠️  Common Mistakes

### ❌ Mistake 1 — Confusing `P(A|B)` with `P(B|A)`

These are generally different.

```text
P(Disease | Positive)
```

is not the same as:

```text
P(Positive | Disease)
```

---

### ❌ Mistake 2 — Ignoring the Prior

The prior probability can strongly influence the posterior.

---

### ❌ Mistake 3 — Assuming Test Accuracy Equals Posterior Probability

A test's sensitivity or specificity does not automatically equal:

```text
P(Disease | Positive)
```

---

### ❌ Mistake 4 — Forgetting False Positives

The evidence probability must account for all ways the evidence can occur.

---

### ❌ Mistake 5 — Ignoring Base Rates

Rare events can have surprisingly low posterior probabilities even after positive evidence.

# 💎  Key Insights

- 🧠 Bayes' Theorem updates probabilities using new evidence.
- 🟡 Prior = initial belief.
- 🔵 Likelihood = probability of evidence under the hypothesis.
- 🟣 Evidence = probability of observing the evidence.
- 🟢 Posterior = updated probability.
- 🔗 Bayes' Theorem comes directly from Conditional Probability.
- ⚠️ `P(A | B)` is generally not equal to `P(B | A)`.
- 🏥 Medical diagnosis is a classic application.
- 📧 Spam detection is another important application.
- 🤖 Naive Bayes uses Bayes' Theorem for classification.
- 📊 Base rates can strongly affect posterior probabilities.
- 🧠 New evidence can update our beliefs.

### ⭐ Remember

```text
Prior
  +
Evidence
  ↓
Posterior
```

Or mathematically:

```text
Posterior
=
Likelihood × Prior
──────────────────
Evidence
```

# 📌  Summary

In this notebook, we learned **Bayes' Theorem**.

The central formula is:

```text
P(A | B)
=
P(B | A)P(A)
────────────
P(B)
```

We learned that:

```text
P(A)
```

is the prior probability.

```text
P(B | A)
```

is the likelihood.

```text
P(B)
```

is the evidence.

```text
P(A | B)
```

is the posterior probability.

The most important conceptual idea is:

> **Bayes' Theorem allows us to update a probability when new evidence becomes available.**
