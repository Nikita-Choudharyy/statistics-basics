# 🧠 What is Probability?

**Probability** is a numerical measure of how likely an event is to occur.

Its value lies between:

```text
0 and 1
```

where:

```text
0 → Impossible Event
1 → Certain Event
```

### 💡 Simple Interpretation

```text
Probability = 0
        ↓
Impossible

Probability = 0.5
        ↓
Equal likelihood

Probability = 1
        ↓
Certain
```

For example, when tossing a fair coin:

```text
P(Heads) = 0.5
P(Tails) = 0.5
```

---

# 📊 Probability Scale

Probability can be represented using a scale.

```text
0                         0.5                         1
│--------------------------│--------------------------│
Impossible             Equally Likely              Certain
```

We can also express probability as a percentage.

```text
0     = 0%
0.25  = 25%
0.50  = 50%
0.75  = 75%
1.00  = 100%
```

### 🔑 Important

> **Probability tells us how likely an event is, not what will definitely happen in one trial.**

For example:

```text
P(Heads) = 0.5
```

does not mean that every two coin flips will contain exactly one Head.

It means that over many repeated trials, the proportion of Heads tends to approach 0.5 for a fair coin.

# 🎲 Random Experiment

A **Random Experiment** is a process whose exact outcome cannot be predicted with certainty before it is performed.

### Examples

- 🪙 Tossing a coin
- 🎲 Rolling a die
- 🃏 Drawing a card
- 🎯 Selecting a random student
- 🏭 Inspecting a randomly selected product

### Example — Coin Toss

Suppose we toss a coin.

Before tossing it, we do not know whether the result will be:

```text
Heads
```

or:

```text
Tails
```

Therefore, tossing a coin is a **random experiment**.

# 🌌 Sample Space

The **Sample Space** is the set of **all possible outcomes** of a random experiment.

It is usually represented by:

```text
S
```

### 🪙 Coin Toss

For one coin toss:

```text
S = {H, T}
```

where:

```text
H = Heads
T = Tails
```

Therefore, the sample space contains two possible outcomes.

### 🎲 Rolling a Die

For one roll of a six-sided die:

```text
S = {1, 2, 3, 4, 5, 6}
```

There are six possible outcomes.



### 🔑 Important

> **Sample Space = Set of all possible outcomes.**

---

# 🔵 Event

An **Event** is a collection of one or more outcomes from the sample space.

An event is usually represented by a capital letter such as:

```text
A
B
C
```

### Example — Die Roll

Sample space:

```text
S = {1, 2, 3, 4, 5, 6}
```

Suppose Event A is:

> Getting an even number.

Then:

```text
A = {2, 4, 6}
```

Event A contains three outcomes.

---

# 🧩 Outcome vs Event

It is important to understand the difference.

### Outcome

A single possible result.

```text
4
```

### Event

A collection of possible outcomes.

```text
{2, 4, 6}
```

For example:

```text
Experiment → Roll a die

Outcome → 4

Event → Getting an even number
       → {2, 4, 6}
```

### 🔑 Remember

> **Outcome = One possible result**

> **Event = One or more outcomes grouped together**

# 🧮 8. Basic Probability Formula

When all outcomes are **equally likely**, the probability of an event can be calculated as:

```text
P(A) = Number of Favorable Outcomes / Total Number of Possible Outcomes
```

Where:

- **P(A)** = Probability of Event A
- **Favorable Outcomes** = Outcomes that satisfy the event
- **Total Outcomes** = All possible outcomes

 --- 

# 🎲 Example — Probability of Rolling a 4

Consider a fair six-sided die.

Sample space:

```text
S = {1, 2, 3, 4, 5, 6}
```

We want:

```text
A = Getting a 4
```

There is:

```text
1 favorable outcome
```

and:

```text
6 total outcomes
```

Therefore:

```text
P(A) = 1 / 6
```

Approximately:

```text
P(A) ≈ 0.1667
```

As a percentage:

```text
≈ 16.67%
```

---

# 🎲 Example — Probability of an Even Number

For a six-sided die:

```text
S = {1, 2, 3, 4, 5, 6}
```

Even outcomes:

```text
A = {2, 4, 6}
```

Number of favorable outcomes:

```text
3
```

Total outcomes:

```text
6
```

Therefore:

```text
P(A) = 3 / 6
     = 0.5
```

So:

```text
P(A) = 50%
```

---

# 🪙 Example — Coin Toss

For a fair coin:

```text
S = {H, T}
```

Probability of Heads:

```text
P(H) = 1 / 2
     = 0.5
```

Probability of Tails:

```text
P(T) = 1 / 2
     = 0.5
```

Therefore:

```text
P(H) + P(T) = 1
```

This makes sense because one of the two outcomes must occur.

# ⚖️ Probability of Impossible and Certain Events

Probability has two important extreme cases.

### ❌ Impossible Event

An event that cannot occur.

Example:

> Rolling a 7 on a standard six-sided die.

The sample space is:

```text
{1, 2, 3, 4, 5, 6}
```

7 is not possible.

Therefore:

```text
P(Rolling 7) = 0
```

### ✅ Certain Event

An event that must occur.

Example:

> Rolling a number less than 7 on a standard six-sided die.

All outcomes satisfy this condition:

```text
{1, 2, 3, 4, 5, 6}
```

Therefore:

```text
P(Number < 7) = 1
```
 ---

# 📌 Basic Properties of Probability

Probability follows some fundamental rules.

### Rule 1 — Probability Range

```text
0 ≤ P(A) ≤ 1
```

Probability can never be less than 0 or greater than 1.

---

### Rule 2 — Certain Event

```text
P(S) = 1
```

The probability of the entire sample space is 1.

---

### Rule 3 — Impossible Event

```text
P(∅) = 0
```

The empty event has probability 0.

---

### Rule 4 — Complement

For an event A:

```text
P(Aᶜ) = 1 − P(A)
```

where:

```text
Aᶜ = Event A does not occur
```

---


# 🔄 Complement of an Event

The **Complement** of an event means the event does not happen.

Suppose:

```text
A = Getting Heads
```

Then:

```text
Aᶜ = Getting Tails
```

If:

```text
P(A) = 0.5
```

then:

```text
P(Aᶜ) = 1 − 0.5
       = 0.5
```

### 💡 Example

Suppose:

```text
P(Rain) = 0.30
```

Then the probability of:

```text
No Rain
```

is:

```text
P(No Rain) = 1 − 0.30
           = 0.70
```

So:

```text
P(No Rain) = 70%
```

# 🧠 Theoretical Probability

**Theoretical Probability** is calculated using mathematical reasoning based on the possible outcomes.

For equally likely outcomes:

```text
P(A) = Favorable Outcomes / Total Outcomes
```

### Example

For a fair die:

```text
P(6) = 1/6
```

This value comes from the mathematical structure of the experiment.

---

# 🧪 Experimental Probability

**Experimental Probability** is calculated from actual observations or repeated experiments.

### Formula

```text
Experimental Probability
=
Number of Times Event Occurred
/
Number of Trials
```

### Example

Suppose we toss a coin 100 times and observe:

```text
Heads = 48
Tails = 52
```

Experimental probability of Heads:

```text
P(H) = 48 / 100
     = 0.48
```

So:

```text
Experimental P(H) = 48%
```

The theoretical probability is:

```text
0.50
```

The experimental result is:

```text
0.48
```

They are close, but not exactly equal.

---


# 📊 Theoretical vs Experimental Probability

| 📐 Theoretical Probability | 🧪 Experimental Probability |
|---|---|
| Based on mathematical reasoning | Based on actual observations |
| Uses possible outcomes | Uses observed outcomes |
| Does not require repeated trials | Requires experimental data |
| Example: Fair coin P(H) = 0.5 | Example: 48 Heads out of 100 = 0.48 |

### 🔑 Important

As the number of repeated trials becomes very large, experimental probability tends to move closer to theoretical probability under stable conditions.

This idea is closely related to the **Law of Large Numbers**, which we will explore further later.

# 🐍 Simulating a Coin Toss Using Python

We can use Python to simulate random experiments.


````python
import numpy as np

result = np.random.choice(["Heads", "Tails"])

print("Result:", result)
````

Every time you run the code, the result may be different.

Example:

```text
Result: Heads
```

or:

```text
Result: Tails
```

This demonstrates randomness.

---

# 🪙 Simulating Multiple Coin Tosses

Let's simulate 10 coin tosses.

````python
results = np.random.choice(
    ["Heads", "Tails"],
    size=10
)

print(results)
````

# 📊 Experimental Probability Using Python

Let's simulate 1,000 coin tosses.

````python
results = np.random.choice(
    ["Heads", "Tails"],
    size=1000
)

heads_count = np.sum(results == "Heads")

experimental_probability = heads_count / len(results)

print("Heads:", heads_count)
print("Experimental Probability of Heads:", experimental_probability)
````

# 📈 Probability and Number of Trials

Let's compare different numbers of coin tosses.

````python
trial_sizes = [10, 100, 1000, 10000]

for trials in trial_sizes:
    results = np.random.choice(
        ["Heads", "Tails"],
        size=trials
    )

    heads_count = np.sum(results == "Heads")
    probability = heads_count / trials

    print(
        f"Trials: {trials:>5} | "
        f"Heads Probability: {probability:.4f}"
    )
````

### 👀 Observation

With a small number of trials, the experimental probability can fluctuate considerably.

As the number of trials increases, the observed proportion generally gets closer to:

```text
0.5
```

### 🔑 Key Idea

> **More repeated observations generally give us a more stable estimate of the underlying probability.**

---

# 📊  Visualizing Experimental Probability

Let's visualize how the estimated probability of Heads changes as the number of tosses increases.


````python
import matplotlib.pyplot as plt

trials = 1000

results = np.random.choice(
    [0, 1],
    size=trials
)

cumulative_probability = np.cumsum(results) / np.arange(1, trials + 1)

plt.plot(cumulative_probability)

plt.axhline(
    0.5,
    linestyle="--",
    label="Theoretical Probability = 0.5"
)

plt.title("Experimental Probability of Heads")
plt.xlabel("Number of Tosses")
plt.ylabel("Cumulative Probability")
plt.legend()

plt.show()
````

### 👀 Observation

The cumulative probability fluctuates initially.

As the number of tosses increases, it tends to stabilize around:

```text
0.5
```

# 🎲 Simulating Dice Rolls

We can also simulate rolling a die.

````python
rolls = np.random.randint(
    1,
    7,
    size=20
)

print(rolls)
````

# 📊 Experimental Probability of Rolling a 6

Let's roll the die 10,000 times.

````python
rolls = np.random.randint(
    1,
    7,
    size=10000
)

six_count = np.sum(rolls == 6)

experimental_probability = six_count / len(rolls)

theoretical_probability = 1 / 6

print("Experimental Probability:", experimental_probability)
print("Theoretical Probability:", theoretical_probability)
````

# 📈 Visualizing Dice Outcomes

````python
values, counts = np.unique(
    rolls,
    return_counts=True
)

plt.bar(values, counts)

plt.title("Frequency of Dice Rolls")
plt.xlabel("Dice Value")
plt.ylabel("Frequency")

plt.show()
````


### 👀 Observation

For a fair die and a sufficiently large number of trials, the frequencies of the six outcomes should become relatively similar.

They will not necessarily be exactly equal.

# 🧠 Probability vs Certainty

It is important not to confuse probability with certainty.

Suppose:

```text
P(Rain) = 0.90
```

This means:

> Rain is highly likely.

It does **not** mean:

> Rain is guaranteed.

Similarly:

```text
P(Rain) = 0.10
```

means:

> Rain is unlikely.

It does not mean:

> Rain is impossible.

### 🔑 Remember

```text
Probability → Likelihood
Certainty   → Guaranteed outcome
```
