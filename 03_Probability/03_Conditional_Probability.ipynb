# 🧠 What is Conditional Probability?

Suppose we want to calculate:

> What is the probability that a student knows Python?

Normally, we might consider the entire group of students.

But suppose we now know:

> The student already knows NumPy.

Now our question becomes:

> **What is the probability that the student knows Python, given that the student knows NumPy?**

This is Conditional Probability.

We are no longer considering everyone.

We are considering only the students who satisfy the given condition.

---

# 🔑 The Meaning of "Given That"

Conditional Probability is often expressed using the phrase:

```text
"given that"
```

For example:

```text
Probability of A given B
```

is written as:

```text
P(A | B)
```

The vertical bar:

```text
|
```

is read as:

```text
"given"
```

Therefore:

```text
P(A | B)
```

means:

> **Probability of A given B**

---

# 🎲 Simple Dice Example

Consider a fair six-sided die.

The sample space is:

```text
S = {1, 2, 3, 4, 5, 6}
```

Suppose:

```text
A = Getting an even number
```

Therefore:

```text
A = {2, 4, 6}
```

Normally:

```text
P(A) = 3/6 = 1/2
```

Now suppose we are told:

> The number rolled is greater than 3.

This is our condition.

Let:

```text
B = Getting a number greater than 3
```

Therefore:

```text
B = {4, 5, 6}
```

Now the question becomes:

> What is the probability of getting an even number **given that** the number is greater than 3?

In notation:

```text
P(A | B)
```

---


# 🧠 Restricting the Sample Space

This is the most important intuition behind Conditional Probability.

Originally:

```text
S = {1, 2, 3, 4, 5, 6}
```

But we already know:

```text
B = {4, 5, 6}
```

Therefore, we no longer need to consider:

```text
{1, 2, 3}
```

Our new relevant sample space becomes:

```text
{4, 5, 6}
```

Among these values:

```text
4 → Even
5 → Odd
6 → Even
```

Therefore, favorable outcomes are:

```text
{4, 6}
```

There are:

```text
2 favorable outcomes
```

and:

```text
3 possible outcomes under the condition
```

Therefore:

```text
P(A | B) = 2/3
```

### 💡 Key Intuition

> **Conditional Probability changes the set of outcomes we are considering because we already know something.**

---

# 📐 Conditional Probability Formula

The mathematical definition of Conditional Probability is:



Where:

- `P(A | B)` → Probability of A given B
- `P(A ∩ B)` → Probability that A and B both occur
- `P(B)` → Probability that B occurs

### 🔑 Important

The condition:

```text
B
```

must have a non-zero probability.

That is:

```text
P(B) > 0
```

---


# 🧮 6. Understanding the Formula Intuitively

The formula:

```text
P(A | B) = P(A ∩ B) / P(B)
```

can be understood as:

```text
Probability of both A and B
────────────────────────────
Probability of the condition B
```

Why?

Because once we know that B has happened, we are effectively asking:

> **Out of the B cases, how many are also A cases?**

This is exactly the idea of conditional probability.

---


# 🎲 Solving the Dice Example Using the Formula

Recall:

```text
A = Even number = {2, 4, 6}

B = Number greater than 3 = {4, 5, 6}
```

Their intersection is:

```text
A ∩ B = {4, 6}
```

Therefore:

```text
P(A ∩ B) = 2/6
```

And:

```text
P(B) = 3/6
```

Using the formula:

```text
P(A | B) = P(A ∩ B) / P(B)
```

```text
= (2/6) / (3/6)
```

```text
= 2/3
```

Therefore:

```text
P(A | B) = 2/3
```

---

# 🧠 Why Does the Denominator Change?

This is one of the most common points of confusion.

For ordinary probability:

```text
P(A)
```

we consider the entire sample space.

For:

```text
P(A | B)
```

we already know that B occurred.

Therefore, the relevant population becomes:

```text
B
```

So the denominator becomes:

```text
P(B)
```

### Visual Intuition

```text
Original Sample Space
┌───────────────────────────────┐
│                               │
│       ┌───────────┐           │
│       │     B     │           │
│       │   ┌───┐   │           │
│       │   │A∩B│   │           │
│       │   └───┘   │           │
│       └───────────┘           │
│                               │
└───────────────────────────────┘
```

When we know B:

```text
Focus only on B
```

Then ask:

```text
How much of B is also A?
```

That is:

```text
P(A | B)
```

---


# 🔗 Conditional Probability and Intersection

Conditional Probability connects directly to Intersection Probability.

Starting from:

```text
P(A | B) = P(A ∩ B) / P(B)
```

Multiply both sides by `P(B)`:

```text
P(A | B) × P(B) = P(A ∩ B)
```

Therefore:

```text
P(A ∩ B) = P(A | B)P(B)
```

This is an important form of the **Multiplication Rule**.

---

# ✖️  Multiplication Rule

For two events:

```text
P(A ∩ B) = P(A | B)P(B)
```

We can also write:

```text
P(A ∩ B) = P(B | A)P(A)
```

Both are valid.

### 🔑 Important

The conditional probability tells us how likely one event is **after knowing another event occurred**.

---

# 🔄  P(A|B) vs P(B|A)

This is extremely important.

In general:

```text
P(A | B) ≠ P(B | A)
```

These are different probabilities.

### `P(A | B)`

Means:

> Probability of A given B.

### `P(B | A)`

Means:

> Probability of B given A.

The condition changes.

# 🐼 Conditional Probability with Pandas

Let's create a small dataset.

````python
import pandas as pd

df = pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Python": [1, 1, 1, 0, 1, 0, 0, 1],
    "SQL":    [1, 1, 0, 1, 1, 0, 1, 0]
})

df
````

# 🔍 Filtering Based on the Condition

Suppose we want:

```text
P(Python | SQL)
```

First, filter students who know SQL.

````python
sql_students = df[df["SQL"] == 1]

sql_students
````

# 🧮 Calculating Conditional Probability

Among SQL students, count how many also know Python.

````python
python_and_sql = (
    (df["Python"] == 1) &
    (df["SQL"] == 1)
).sum()

sql_count = (
    df["SQL"] == 1
).sum()

conditional_probability = (
    python_and_sql / sql_count
)

print("P(Python | SQL):", conditional_probability)
````

# 📊 Conditional Probability Using a Contingency Table

A **Contingency Table** helps us organize relationships between categorical variables.

````python
pd.crosstab(
    df["Python"],
    df["SQL"]
)
````

This table shows how Python and SQL knowledge are distributed together.

Contingency tables are very useful for:

- 📊 Categorical Data Analysis
- 🔍 Association Analysis
- 🧪 Chi-Square Testing
- 🤖 Feature Analysis

# 🧮 Using a Larger Example

Let's create a simple customer dataset.

````python
customers = pd.DataFrame({
    "Customer": range(1, 11),
    "Purchased_Laptop": [1, 1, 0, 1, 0, 1, 0, 0, 1, 0],
    "Purchased_Phone":  [1, 0, 1, 1, 0, 1, 1, 0, 0, 1]
})

customers
````

# 🔍  Filtering Customers

First find customers who purchased a phone.

````python
phone_customers = customers[
    customers["Purchased_Phone"] == 1
]

phone_customers
````

# 🧮 Calculating the Conditional Probability

````python
phone_count = (
    customers["Purchased_Phone"] == 1
).sum()

laptop_and_phone = (
    (customers["Purchased_Laptop"] == 1) &
    (customers["Purchased_Phone"] == 1)
).sum()

p_laptop_given_phone = (
    laptop_and_phone / phone_count
)

print(
    "P(Laptop | Phone):",
    p_laptop_given_phone
)
````

### 💡 Interpretation

The result tells us:

> Among customers who purchased a phone, what proportion also purchased a laptop?

This is a very useful way to think about conditional relationships in real datasets.

# 🎲 Conditional Probability with Dice in Python

Let's verify the earlier dice example.

````python
import numpy as np

sample_space = np.arange(1, 7)

A = sample_space % 2 == 0
B = sample_space > 3

intersection = A & B

p_A_and_B = np.mean(intersection)
p_B = np.mean(B)

p_A_given_B = p_A_and_B / p_B

print("P(A ∩ B):", p_A_and_B)
print("P(B):", p_B)
print("P(A | B):", p_A_given_B)
````

# 🎲 Conditional Probability Through Simulation

We can also simulate many dice rolls.

````python
import numpy as np

rolls = np.random.randint(
    1,
    7,
    size=100000
)

condition = rolls > 3

filtered_rolls = rolls[condition]

event_A = filtered_rolls % 2 == 0

conditional_probability = np.mean(event_A)

print(
    "Experimental P(Even | Number > 3):",
    conditional_probability
)
````

# 📊 Visualizing Conditional Probability

We can visualize the filtered outcomes.

````python
import matplotlib.pyplot as plt

unique_values, counts = np.unique(
    filtered_rolls,
    return_counts=True
)

plt.bar(unique_values, counts)

plt.title(
    "Dice Outcomes Given Number > 3"
)

plt.xlabel("Dice Value")
plt.ylabel("Frequency")

plt.show()
````

### 👀 Observation

Because we already know:

```text
Roll > 3
```

the visualization contains only:

```text

```

This demonstrates the key idea:

> **Conditional Probability restricts the sample space based on known information.**

# 🔄  Conditional Probability and Independence

Two events A and B are independent when knowing B does not change the probability of A.

That means:

```text
P(A | B) = P(A)
```

Similarly:

```text
P(B | A) = P(B)
```

### Example — Coin Tosses

Suppose we toss a coin twice.

Let:

```text
A = Heads on first toss
B = Heads on second toss
```

The result of the second toss is not affected by the first toss.

Therefore:

```text
P(A | B) = P(A)
```

and:

```text
P(B | A) = P(B)
```

---

# 🧠  Important Relationship

For independent events:

```text
P(A | B) = P(A)
```

Using the conditional probability formula:

```text
P(A | B) = P(A ∩ B) / P(B)
```

If:

```text
P(A | B) = P(A)
```

then:

```text
P(A ∩ B) = P(A)P(B)
```

Therefore:

```text
Independent Events
        ↓
P(A | B) = P(A)
        ↓
P(A ∩ B) = P(A)P(B)
```

This connects Conditional Probability directly to the Multiplication Rule.

---

# ⚠️ Conditional Probability Does Not Mean Causation

This is an important statistical idea.

Suppose:

```text
P(A | B)
```

is high.

It does **not automatically mean**:

```text
B causes A
```

It only tells us:

> The probability of A when B is known to be true.

### Example

Suppose:

```text
P(Ice Cream Sales are High | Temperature is High)
```

is high.

This tells us that high ice cream sales are more common when temperatures are high.

But we need additional evidence before making causal conclusions.

### 🔑 Remember

```text
Conditional Relationship ≠ Causation
```

# 📊  Conditional Probability vs Ordinary Probability

| 📊 Ordinary Probability | 🔗 Conditional Probability |
|---|---|
| `P(A)` | `P(A \| B)` |
| No additional condition | Condition B is known |
| Uses entire relevant sample space | Focuses on B |
| Example: P(Even) | Example: P(Even \| Number > 3) |

### Example

Ordinary:

```text
P(Even) = 3/6 = 0.5
```

Conditional:

```text
P(Even | Number > 3) = 2/3
```

The probability changed because we received additional information.

# 📌  Summary

In this notebook, we learned **Conditional Probability**.

The central idea is:

> **We calculate the probability of an event after receiving information about another event.**

The notation is:

```text
P(A | B)
```

which means:

```text
Probability of A given B
```

The main formula is:

```text
P(A | B) = P(A ∩ B) / P(B)
```

We also learned:

```text
P(A ∩ B) = P(A | B)P(B)
```

For independent events:

```text
P(A | B) = P(A)
```

And we learned why:

```text
P(A | B)
```

and:

```text
P(B | A)
```

are generally different.

---


# 🚀 Final Takeaway

Conditional Probability teaches us one powerful idea:

> **Probability can change when we receive new information.**

Think of it like:

```text
Initial Information
        ↓
     P(A)
        ↓
New Information B
        ↓
    P(A | B)
        ↓
Updated Probability
```

This way of thinking is fundamental to:

- 📊 Statistics
- 🧠 Bayesian Reasoning
- 🤖 Machine Learning
- 🎯 Classification
- 🏥 Medical Diagnosis
- 📧 Spam Detection
- 🛒 Recommendation Systems
