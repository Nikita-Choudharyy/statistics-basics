# 🧠  What is a Population?

In Statistics, a **population** means:

> **The complete group of individuals, observations, objects, or measurements that we are interested in studying.**

The population does not necessarily mean people.

It can be:

- 👨‍🎓 All students in a university
- 🛒 All customers of a company
- 🚗 All cars produced by a manufacturer
- 🏥 All patients in a hospital
- 📦 All products manufactured in a factory
- 💰 All transactions made by a business
- 🌍 All people in a country

Suppose a university has:

```text

```

If we want to study the average age of all students, then:

```text

```

But this may require:

```text

Resources
```

This is where sampling becomes useful.

# 🔬  What is a Sample?

> **A smaller subset selected from the population for analysis.**

Suppose the university has:

```text

```

The sample is a smaller part of the population.

# 📐  Population Mean

The population mean is:

```text
μ = ΣX / N
```

Where:

```text
μ = Population Mean

ΣX = Sum of all population values

N = Population Size
```

---

# 🧮  Example of Population Mean

Suppose the population is:

```text
[10, 20, 30, 40, 50]
```

Then:

```text
μ = (10 + 20 + 30 + 40 + 50) / 5
```

Therefore:

```text
μ = 30
```

---

# 🐍  Python Implementation


````python
import numpy as np

population = np.array([
    10, 20, 30, 40, 50
])

population_mean = np.mean(
    population
)

print("Population Mean:", population_mean)
````

# 🔬  Sample Mean

The sample mean is:

```text
x̄ = Σx / n
```

Where:

```text
x̄ = Sample Mean

Σx = Sum of sample values

n = Sample Size
```

Notice the difference:

```text
Population → N
Sample     → n
```

---

# 🐍  Sample Mean Example

````python
sample = np.array([
    20, 30, 40
])

sample_mean = np.mean(
    sample
)

print("Sample Mean:", sample_mean)
````

# 📊  Population and Sample Variance

The formulas are slightly different.

### Population Variance

```text
σ² = Σ(X - μ)² / N
```

### Sample Variance

```text
s² = Σ(x - x̄)² / (n - 1)
```

The sample variance uses:

```text
n - 1
```

This is called **Bessel's correction**.

---

# 🧠  Why n - 1?

When we calculate population variance, we know the actual population mean.

But when working with a sample, we estimate the population mean using the sample mean.

This introduces a degree of freedom adjustment.

Therefore, sample variance commonly uses:

```text
n - 1
```

instead of:

```text
n
```

This helps produce an unbiased estimator of the population variance under standard assumptions.

---

# 🐍  Population vs Sample Variance in Python

NumPy allows us to specify the degrees of freedom.


````python
data = np.array([
    10, 20, 30, 40, 50
])

population_variance = np.var(
    data
)

sample_variance = np.var(
    data,
    ddof=1
)

print(
    "Population Variance:",
    population_variance
)

print(
    "Sample Variance:",
    sample_variance
)
````

# 📊  Population vs Sample Standard Deviation

Similarly:

````python
population_std = np.std(
    data
)

sample_std = np.std(
    data,
    ddof=1
)

print(
    "Population Standard Deviation:",
    population_std
)

print(
    "Sample Standard Deviation:",
    sample_std
)
````

# 🛒  Example with Customer Spending

Let's create a hypothetical population.


````python
np.random.seed(42)

customer_spending = np.random.normal(
    loc=3000,
    scale=800,
    size=100000
)

customer_spending[:10]
````

# 📊  Calculate Population Mean

````python
population_mean = np.mean(
    customer_spending
)

print(
    "Population Mean:",
    population_mean
)
````

# 🔬 Select a Sample

Suppose we select:

```text
Sample Size = 500
```

using random sampling.

````python
sample = np.random.choice(
    customer_spending,
    size=500,
    replace=False
)

sample[:10]
````

# 📊  Calculate Sample Mean


````python
sample_mean = np.mean(
    sample
)

print(
    "Population Mean:",
    population_mean
)

print(
    "Sample Mean:",
    sample_mean
)
````

# 🧠  Why Are They Different?

The sample is only a subset of the population.

Therefore:

```text
Sample
   ≠
Entire Population
```

The sample mean can vary depending on which observations were selected.

This difference is called:

### ⚠️ Sampling Error

---

# 📐  Sampling Error

Sampling error is the difference between a sample statistic and the corresponding population parameter.

For the mean:

```text
Sampling Error
=
Sample Mean - Population Mean
```

Mathematically:

```text
Sampling Error = x̄ - μ
```

---

# 🐍 Calculate Sampling Error

````python
sampling_error = (
    sample_mean - population_mean
)

print(
    "Sampling Error:",
    sampling_error
)
````

# 🎯  Sampling Variability

If we repeatedly select samples from the same population, their statistics will not always be identical.

For example:

```text
Sample 1 → Mean = 2992
Sample 2 → Mean = 3011
Sample 3 → Mean = 2987
Sample 4 → Mean = 3005
Sample 5 → Mean = 2998
```

This variation is called:

### 📊 Sampling Variability

---

# 🧠 Demonstrating Sampling Variability

Let's select multiple samples.

````python
sample_means = []

for _ in range(100):
    sample = np.random.choice(
        customer_spending,
        size=500,
        replace=False
    )

    sample_means.append(
        np.mean(sample)
    )

sample_means[:10]
````

# 📊  Visualize Sample Means


````python
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(
    sample_means,
    bins=20,
    kde=True
)

plt.axvline(
    population_mean,
    linestyle="--",
    label="Population Mean"
)

plt.title(
    "Distribution of Sample Means"
)

plt.xlabel(
    "Sample Mean"
)

plt.ylabel(
    "Frequency"
)

plt.legend()

plt.show()
````

### 👀 Observation

The sample means are generally centered around the population mean.

This idea becomes extremely important when we study:

```text
📊 Sampling Distributions
```

and later:

```text
🧠 Central Limit Theorem
```

---

# 🔢  Effect of Sample Size

Sample size matters.

Let's compare:

```text
Small Sample
```

with:

```text
Large Sample
```

We can investigate how much sample means vary.

---

# 🐍  Small Sample

````python
small_sample_means = []

for _ in range(500):
    sample = np.random.choice(
        customer_spending,
        size=20,
        replace=False
    )

    small_sample_means.append(
        np.mean(sample)
    )

````

# 🐍  Large Sample


````python
large_sample_means = []

for _ in range(500):
    sample = np.random.choice(
        customer_spending,
        size=500,
        replace=False
    )

    large_sample_means.append(
        np.mean(sample)
    )
````

# 📊  Compare Variability


````python
print(
    "Small Sample Mean Standard Deviation:",
    np.std(small_sample_means)
)

print(
    "Large Sample Mean Standard Deviation:",
    np.std(large_sample_means)
)
````

Typically:

```text
Small Sample
→ More variation
```

```text
Large Sample
→ Less variation
```

This is an important statistical idea.

---

# 🧠  Why Larger Samples Help

As sample size increases:

```text
Sample Size ↑
      ↓
Sampling Variability ↓
      ↓
Sample Statistics become more stable
```

This is one reason larger, representative samples are often useful.

However:

> A large sample does not automatically eliminate bias.

If the sampling method is biased, increasing the sample size may simply give us a very precise estimate of the wrong thing.

---

# ⚠️  Sampling Error vs Sampling Bias

These concepts are different.

### Sampling Error

Occurs because:

```text
Sample ≠ Entire Population
```

It is a natural consequence of sampling.



### Sampling Bias

Occurs when the sampling process systematically favors certain members or outcomes.

For example:

```text
Population
100,000 Customers

Sample
Only High-Spending Customers
```

This sample may not represent the entire customer population.

---

# 🧠  Representative Sample

A good sample should represent important characteristics of the population.

Suppose the population contains:

```text
50% Group A
30% Group B
20% Group C
```

A sample that roughly reflects these characteristics may be more representative than a sample containing only Group A.

The exact strategy depends on the problem and sampling design.

---

# 🎯  What Makes a Good Sample?

A useful sample should ideally be:

- 🎯 Relevant to the research question
- 🔀 Selected using an appropriate method
- 🌍 Representative of the target population
- 📊 Large enough for the intended analysis
- ⚖️ Minimally biased
- 🔬 Collected consistently

# 🔬  Random Sampling

Random sampling gives population members a known chance of being selected under the chosen sampling design.

The simplest form is:

### 🎲 Simple Random Sampling

Every population member has an equal probability of selection.

---

# 🐍 Simple Random Sampling with NumPy

````python
population = np.arange(
    1,
    101
)

sample = np.random.choice(
    population,
    size=10,
    replace=False
)

print(
    "Population:",
    population
)

print(
    "Random Sample:",
    sample
)
````

# 📊  Understanding the Example

Here:

```text
Population = 1 to 100
```

We selected:

```text
Sample Size = 10
```

without replacement.

Therefore:

```text
Each selected value appears only once.
```

---

# 🔄  Sampling With Replacement

When:

```text
replace=True
```

the same observation can be selected more than once.

````python
sample_with_replacement = np.random.choice(
    population,
    size=10,
    replace=True
)

print(
    sample_with_replacement
)
````

# 🚫  Sampling Without Replacement

When:

```text
replace=False
```

an observation cannot be selected again after it has been chosen

````python
sample_without_replacement = np.random.choice(
    population,
    size=10,
    replace=False
)

print(
    sample_without_replacement
)
````

# 🔄  With Replacement vs Without Replacement

| 🔄 With Replacement | 🚫 Without Replacement |
|---|---|
| Observation can be selected again | Observation cannot be selected again |
| Duplicates possible | No duplicate selected observations |
| Useful in some simulation methods | Common for selecting a subset |
| Population effectively remains available | Population pool decreases after each selection |

---

# 🧠  Sampling Fraction

The sampling fraction is:

```text
Sample Size
──────────────
Population Size
```

For example:

```text
Population = 100,000
Sample = 1,000
```

Then:

```text
Sampling Fraction = 1000 / 100000
                  = 0.01
                  = 1%
```

---

# 🐍  Calculate Sampling Fraction

````python
population_size = 100000
sample_size = 1000

sampling_fraction = (
    sample_size / population_size
)

print(
    "Sampling Fraction:",
    sampling_fraction
)

print(
    "Sampling Percentage:",
    sampling_fraction * 100,
    "%"
)
````

# 🧪  Python Experiment — Repeated Sampling

Let's perform a larger simulation.

````python
np.random.seed(42)

population = np.random.normal(
    loc=100,
    scale=15,
    size=100000
)

population_mean = np.mean(
    population
)

sample_means = []

for _ in range(1000):
    sample = np.random.choice(
        population,
        size=100,
        replace=False
    )

    sample_means.append(
        np.mean(sample)
    )
print(
    "Population Mean:",
    population_mean
)

print(
    "Average of Sample Means:",
    np.mean(sample_means)
)
````

# 📊  Visualize Repeated Sample Means

````python
sns.histplot(
    sample_means,
    bins=30,
    kde=True
)

plt.axvline(
    population_mean,
    linestyle="--",
    label="Population Mean"
)

plt.title(
    "Repeated Sample Means"
)

plt.xlabel(
    "Sample Mean"
)

plt.ylabel(
    "Frequency"
)

plt.legend()

plt.show()
````

# ⚠️  Common Sampling Mistakes

### ❌ Mistake 1: Using a Biased Sample

Example:

```text
Want opinions of all customers
        ↓
Ask only premium customers
```

The sample may not represent the full population.

---

### ❌ Mistake 2: Very Small Sample

A very small sample may produce unstable estimates.

---

### ❌ Mistake 3: Ignoring Important Groups

If an important subgroup is missing, conclusions may be misleading.

---

### ❌ Mistake 4: Assuming Larger Always Means Better

A huge biased sample can still produce biased results.

---

### ❌ Mistake 5: Ignoring Data Dependence

Observations may be related to each other.

Examples:

```text
Repeated measurements from the same person
Multiple transactions from the same customer
Time-series observations
```

These situations may require specialized sampling or analysis methods.

---

```text
Sampling Error

from selecting a sample
```

while:

```text
Sampling Bias
      ↓
Systematic problem
in how the sample was selected
```

A larger sample can reduce random sampling variability, but it does not automatically remove systematic bias.

# 💡  Key Insights

- 👥 **Population** is the complete group we are interested in studying.
- 🔬 **Sample** is a subset selected from the population.
- 📐 **Parameter** describes a population.
- 📊 **Statistic** describes a sample.
- 🏛️ A census studies the entire population.
- 🎯 Sampling studies only part of the population.
- ⚠️ Sampling creates natural sampling variability.
- 📈 Larger samples generally produce less sampling variability.
- ⚠️ Larger samples do not automatically remove sampling bias.
- 🎲 Random sampling helps reduce selection problems under appropriate designs.
- 📊 Sample statistics can be used to estimate population parameters.
- 🧠 Sampling distributions describe how statistics vary across repeated samples.
- 🤖 Sampling is important in Machine Learning for dataset splitting, evaluation, class imbalance, and large-scale data analysis.
