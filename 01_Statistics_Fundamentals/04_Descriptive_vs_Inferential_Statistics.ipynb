# 📊 Descriptive vs Inferential Statistics

Statistics can be broadly divided into two major areas:

1. 📊 Descriptive Statistics
2. 🔬 Inferential Statistics

Both are used to understand data, but they have different purposes.

**Descriptive Statistics** helps us summarize and describe the data we have.

**Inferential Statistics** helps us use sample data to make conclusions or estimates about a larger population.

This notebook explores the difference between these two approaches and their importance in Data Science and Machine Learning.

## 🧠 The Big Picture

Imagine that a company has data about **100,000 customers**.

The company may want to answer two different types of questions.

### 📊 Descriptive Question

> "What does our existing customer data look like?"

We might calculate:

- Average age
- Median spending
- Most common location
- Standard deviation
- Percentiles

This is **Descriptive Statistics**.

---

### 🔬 Inferential Question

Now suppose the company wants to know:

> "What is the average spending of all potential customers in our market?"

It may not have data about every potential customer.

Instead, it could select a sample and use that sample to estimate the larger population.

This is **Inferential Statistics**.

### 🔑 Core Difference

> **Descriptive Statistics describes the data we have.**

> **Inferential Statistics uses data we have to learn about a larger population.**

## 📊 What is Descriptive Statistics?

Descriptive Statistics is used to **organize, summarize, and describe observed data**.

It does not try to make conclusions about data that was not observed.

### Common Descriptive Statistics

#### 📍 Measures of Central Tendency

- Mean
- Median
- Mode

#### 📏 Measures of Dispersion

- Range
- Variance
- Standard Deviation
- Interquartile Range

#### 📊 Other Summaries

- Percentiles
- Quartiles
- Minimum
- Maximum

#### 📈 Visualizations

- Histograms
- Boxplots
- Bar charts
- Scatterplots

### 🔑 Goal

> **Describe what is present in the dataset.**

## 🔍 Example: Student Marks

Suppose we have the marks of 10 students:

`[55, 60, 65, 70, 70, 75, 80, 85, 90, 95]`

We can describe this dataset using:

- Mean
- Median
- Minimum
- Maximum
- Standard deviation
- Range

These values help us understand the dataset without trying to make conclusions about students who are not included in the data.

````python
import numpy as np

marks = np.array([55, 60, 65, 70, 70, 75, 80, 85, 90, 95])

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Range:", np.max(marks) - np.min(marks))
print("Standard Deviation:", np.std(marks))
````

### 💡 Interpretation

The calculated values give us a numerical summary of the marks.

For example:

- 📊 Mean tells us the average marks.
- 📍 Median tells us the middle value.
- 📉 Minimum tells us the lowest observed mark.
- 📈 Maximum tells us the highest observed mark.
- ↔️ Range tells us the difference between the highest and lowest marks.
- 📏 Standard deviation tells us how spread out the marks are.

All of these are examples of **Descriptive Statistics**.

## 📈 Visualizing Descriptive Statistics

Descriptive Statistics is not limited to numerical calculations.

Visualizations can also help us understand the structure of the data.

Let's visualize the distribution of student marks.

````python
import matplotlib.pyplot as plt

plt.hist(marks, bins=5, edgecolor="black")

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
````

### 👀 Observation

The histogram shows how the student marks are distributed.

We can visually inspect:

- 📊 Where most observations are concentrated
- 📈 Whether the data appears symmetric
- 📉 Whether there are unusually low or high values
- 🔍 The overall shape of the distribution

This is another example of descriptive analysis.

## 🔬 What is Inferential Statistics?

Inferential Statistics uses information from a **sample** to make conclusions, estimates, or predictions about a larger **population**.

This is necessary because studying an entire population is often difficult or impossible.

### Example

Suppose a country has:

**Population = 10 million people**

We want to estimate the average income of the entire population.

Instead of collecting income information from all 10 million people, we may select:

**Sample = 5,000 people**

We can calculate the sample's average income and use statistical methods to make an inference about the population.

### 🔑 Goal

> **Use sample information to learn about a population.**

## 👥 Population → Sample → Inference

The basic idea can be represented as:

**👥 Population**

↓  
Select a sample

↓  

**🎯 Sample**

↓  
Calculate statistics

↓  

**📊 Sample Information**

↓

**🔬 Make an Inference**

↓

**👥 Learn about the Population**

This is the fundamental idea behind Inferential Statistics.

## 📐 Parameter vs Statistic

Two important terms in Inferential Statistics are:

### 📐 Parameter

A parameter describes the **entire population**.

Examples:

- Population mean → **μ**
- Population standard deviation → **σ**
- Population proportion → **p**

---

### 📊 Statistic

A statistic describes a **sample**.

Examples:

- Sample mean → **x̄**
- Sample standard deviation → **s**
- Sample proportion → **p̂**

### 🔑 Remember

> **Population → Parameter**

> **Sample → Statistic**

Inferential Statistics often uses a **sample statistic to estimate a population parameter**.

## ⚖️ Descriptive vs Inferential Statistics

| 📊 Descriptive | 🔬 Inferential |
|---|---|
| Describes observed data | Makes conclusions about a population |
| Summarizes data | Generalizes from a sample |
| Mean, Median, Mode | Confidence Intervals |
| Range, Variance, SD | Hypothesis Testing |
| Percentiles, Quartiles | P-values |
| Charts and Graphs | Statistical Tests |
| Focuses on known data | Deals with uncertainty |
| No generalization required | Generalization is the main goal |

### 💡 Simple Memory Trick

**Descriptive → Describe**

**Inferential → Infer**

## 🧪 Common Inferential Statistical Methods

Inferential Statistics includes several important techniques.

### 📏 Estimation

Estimate unknown population parameters using sample data.

Examples:

- Point estimation
- Confidence intervals

### 🧪 Hypothesis Testing

Test claims or assumptions about a population.

Examples:

- Z-test
- T-test
- Chi-Square test
- F-test
- ANOVA

### 📊 Probability

Probability provides the mathematical foundation for reasoning about uncertainty.

### 🔍 Sampling

Sampling methods help us select useful observations from a population.

These topics will be studied in detail in later sections of this repository.

````python
import numpy as np

population = np.arange(1, 1001)

np.random.seed(42)

sample = np.random.choice(
    population,
    size=50,
    replace=False
)

population_mean = np.mean(population)
sample_mean = np.mean(sample)

print("Population Mean:", population_mean)
print("Sample Mean:", sample_mean)
````

### 🔍 What Did We Do?

We created:

- 👥 A population of 1,000 observations
- 🎯 A random sample of 50 observations

Then we calculated:

- 📐 Population mean
- 📊 Sample mean

The sample mean will generally not be exactly equal to the population mean.

This difference occurs because we are working with only a subset of the population.

Inferential Statistics gives us tools to understand and quantify this uncertainty.

## 🎯 Why Does Uncertainty Matter?

When we calculate a value from the entire population, we know the exact result for that population.

But when we use a sample, there is uncertainty.

For example:

Population mean:

**μ = 50**

Sample mean:

**x̄ = 52**

We should not immediately conclude:

> "The population mean is 52."

Instead, we recognize that:

> **52 is an estimate based on the sample.**

Inferential Statistics helps us understand how reliable such estimates are.

This idea leads to important concepts such as:

- Confidence intervals
- Standard error
- Hypothesis testing
- P-values
- Sampling distributions

## 🧩 When Should We Use Descriptive or Inferential Statistics?

### Use 📊 Descriptive Statistics when:

- You want to summarize a dataset.
- You want to understand the distribution.
- You want to calculate averages.
- You want to measure variability.
- You want to visualize the data.
- You are performing initial EDA.

### Use 🔬 Inferential Statistics when:

- You want to make conclusions about a population.
- You are working with a sample.
- You want to estimate a population parameter.
- You want to test a hypothesis.
- You want to quantify uncertainty.
- You want to compare groups statistically.

### 💡 Simple Decision

Ask:

> **"Am I only describing the data I have?"**

→ Descriptive Statistics

Or:

> **"Am I using this data to make a conclusion about something larger?"**

→ Inferential Statistics
