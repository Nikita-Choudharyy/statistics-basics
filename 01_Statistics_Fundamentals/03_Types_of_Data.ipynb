## 🧠 What is Data?

**Data** refers to facts, observations, measurements, or information collected for analysis.

Examples:

- 👤 Age = 25
- 💰 Salary = ₹50,000
- 🎓 Education = Bachelor's
- ⭐ Rating = 4.5
- 🌡️ Temperature = 32°C
- 🛒 Number of purchases = 8
- 🎨 Favorite color = Blue

Each of these represents a piece of information.

However, these values are not all the same type.

For example:

- `Age` is numerical.
- `Education` is categorical.
- `Rating` can be treated as ordered data.
- `Number of purchases` is a count.

Understanding these differences is important before performing statistical analysis.

## 🗃️ Main Classification of Data

Statistical data can broadly be divided into two major categories:

### 1️⃣ Qualitative Data

Also called **Categorical Data**.

It represents categories or qualities rather than numerical measurements.

Examples:

- Gender
- City
- Product category
- Blood group
- Department
- Color

---

### 2️⃣ Quantitative Data

Also called **Numerical Data**.

It represents values that can be measured or counted numerically.

Examples:

- Age
- Height
- Weight
- Salary
- Number of purchases
- Temperature

### 🔑 Simple Rule

> **Qualitative → Categories**

> **Quantitative → Numbers**

## 🔤 Qualitative Data

Qualitative data describes **categories, characteristics, or qualities**.

It does not primarily represent numerical measurements.

### Examples

| Variable | Example Values |
|---|---|
| 🎨 Color | Red, Blue, Green |
| 🏙️ City | Delhi, Jaipur, Mumbai |
| 🩸 Blood Group | A, B, AB, O |
| 📱 Device Type | Mobile, Tablet, Laptop |
| 🛍️ Product Category | Electronics, Clothing, Food |

Although categories may sometimes be represented using numbers, the numbers do not necessarily have mathematical meaning.

For example:

`1 = Male`

`2 = Female`

Here, calculating the average of these codes would not provide meaningful statistical information.

### 🔑 Key Idea

> **Categorical values represent groups rather than measurable quantities.**

## 🏷️ Nominal Data

**Nominal data** consists of categories that have **no natural order or ranking**.

The categories are simply different labels.

### Examples

- 🎨 Colors → Red, Blue, Green
- 🩸 Blood groups → A, B, AB, O
- 🏙️ Cities → Delhi, Jaipur, Mumbai
- 💻 Operating systems → Windows, macOS, Linux
- 🛒 Product categories → Electronics, Clothing, Food

For example:

`Red < Blue < Green`

does not make statistical sense.

There is no natural ranking between these categories.

### 💡 Important

Numbers can be assigned to nominal categories, but those numbers are simply labels.

Example:

```text
1 → Delhi
2 → Mumbai
3 → Jaipur

## 📋 Ordinal Data

**Ordinal data** consists of categories that have a meaningful order or ranking.

However, the difference between the categories is not necessarily equal or measurable.

### Examples

#### ⭐ Customer Satisfaction

```text
Very Unsatisfied
Unsatisfied
Neutral
Satisfied
Very Satisfied

## 🔢 Quantitative Data

Quantitative data represents **numerical quantities** that can be measured or counted.

Examples:

- 👤 Age = 25
- 💰 Salary = ₹60,000
- 📏 Height = 170.5 cm
- ⚖️ Weight = 65 kg
- 🛒 Number of purchases = 12
- 🌡️ Temperature = 30°C

Quantitative data can be divided into:

### 1️⃣ Discrete Data

Values that are generally **countable**.

### 2️⃣ Continuous Data

Values that can be **measured** and can take values within a range.

## 🔢 Discrete Data

**Discrete data** represents countable values.

These values are usually whole numbers because they represent counts.

### Examples

- 👨‍👩‍👧 Number of family members = 5
- 🛒 Number of purchases = 12
- 📦 Number of products = 20
- 🏠 Number of rooms = 4
- 👨‍🎓 Number of students = 50

For example:

```text
0, 1, 2, 3, 4, 5, ...

## 📏 Continuous Data

**Continuous data** represents values that can be measured.

It can take many possible values within a range.

### Examples

- 📏 Height = 170.5 cm
- ⚖️ Weight = 65.72 kg
- 🌡️ Temperature = 32.6°C
- ⏱️ Time = 4.73 seconds
- 📐 Distance = 12.45 km

For example, a person's height could be:

```text
170 cm
170.1 cm
170.15 cm
170.156 cm
...

## ⚖️ Discrete vs Continuous Data

| 🔢 Discrete | 📏 Continuous |
|---|---|
| Counted | Measured |
| Usually whole numbers | Can contain decimals |
| Number of students | Height |
| Number of purchases | Weight |
| Number of cars | Temperature |
| Number of calls | Time |

### 💡 Easy Trick

Ask yourself:

> **"Am I counting it or measuring it?"**

If you are **counting** → Discrete

If you are **measuring** → Continuous

## 📐 Measurement Scales

Another important way to classify data is through **levels of measurement**.

There are four commonly discussed measurement scales:

1. 🏷️ Nominal
2. 📋 Ordinal
3. 📏 Interval
4. ⚖️ Ratio

These scales tell us what kind of mathematical operations and comparisons make sense for a variable.

## 🏷️ Nominal Scale

Nominal data represents categories with **no order**.

### Examples

- Blood group
- Gender
- Country
- Eye color
- Product category

### Characteristics

- ❌ No natural order
- ❌ Differences are not meaningful
- ❌ Arithmetic operations are not meaningful

### Example

```text
Blood Group:

A
B
AB
O

## 📋 Ordinal Scale

Ordinal data contains categories with a **meaningful order**.

### Examples

- Satisfaction levels
- Education levels
- Competition rankings
- Severity levels

Example:

```text
Low → Medium → High

## 🌡️ Interval Scale

Interval data has:

- ✅ Meaningful order
- ✅ Equal intervals between values
- ❌ No true zero point

A classic example is **temperature measured in Celsius**.

Consider:

```text
10°C
20°C
30°C

## ⚖️ Ratio Scale

Ratio data has:

- ✅ Meaningful order
- ✅ Equal intervals
- ✅ A meaningful zero point

Examples:

- Height
- Weight
- Age
- Income
- Distance
- Number of purchases
- Time duration

For example:

```text
10 kg
20 kg
30 kg

## 🧩 Four Levels of Measurement

| Scale | Order | Equal Intervals | True Zero | Example |
|---|---:|---:|---:|---|
| 🏷️ Nominal | ❌ | ❌ | ❌ | Blood Group |
| 📋 Ordinal | ✅ | ❌ | ❌ | Satisfaction Level |
| 🌡️ Interval | ✅ | ✅ | ❌ | Celsius Temperature |
| ⚖️ Ratio | ✅ | ✅ | ✅ | Weight |

### 🧠 Remember the Progression

**Nominal → Ordinal → Interval → Ratio**

As we move from Nominal to Ratio, the amount of mathematical information available generally increases.

## 🔍 Identifying the Statistical Data Types

Let's classify the variables in our dataset.

| Variable | Statistical Type | Explanation |
|---|---|---|
| Age | Quantitative → Ratio | Numerical measurement with meaningful zero |
| City | Qualitative → Nominal | Categories without order |

### 💡 Important

The same dataset can contain **multiple types of data**.

This is normal in real-world Data Science datasets.

## 🤖 Why Does Data Type Matter in Machine Learning?

Identifying the data type is an important step before preprocessing and modeling.

Different variables may require different treatment.

### 🏷️ Categorical Data

Example:

`City = Jaipur, Delhi, Mumbai`

Machine Learning algorithms usually require numerical representations.

We may use techniques such as:

- One-Hot Encoding
- Ordinal Encoding
- Target Encoding in appropriate situations

---

### 🔢 Numerical Data

Numerical features can often be used directly by ML algorithms.

However, we may still need:

- Scaling
- Transformation
- Outlier detection
- Distribution analysis

---

### 📋 Ordinal Data

If categories have a meaningful order:

```text
Low → Medium → High

## 🔗 Statistics → AI/ML Connection

| 📊 Data Type | 🤖 Possible ML/Data Science Use |
|---|---|
| Nominal | One-hot encoding, frequency analysis |
| Ordinal | Ordered encoding, rank-based analysis |
| Discrete | Counts, frequency analysis, feature engineering |
| Continuous | Scaling, transformations, distribution analysis |
| Interval | Numerical analysis with care around zero |
| Ratio | Numerical analysis, scaling, statistical modeling |

### 💡 Real ML Workflow

A simplified workflow is:

**Raw Dataset**

↓

**Identify Data Types**

↓

**Explore Distributions**

↓

**Handle Missing Values**

↓

**Encode Categorical Variables**

↓

**Scale / Transform Numerical Variables When Needed**

↓

**Train ML Model**

Understanding data types is therefore one of the first steps toward building a reliable Machine Learning pipeline.
