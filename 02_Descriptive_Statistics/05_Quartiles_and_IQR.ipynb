# 🧠  What are Quartiles?

**Quartiles** divide an ordered dataset into **four approximately equal parts**.

The three main quartiles are:

- **Q1 → First Quartile**
- **Q2 → Second Quartile**
- **Q3 → Third Quartile**

They help us understand how values are distributed across the dataset.

### 💡 Simple Idea

```text
0%        25%        50%        75%        100%
│----------│----------│----------│----------│
           Q1         Q2         Q3
```

Therefore:

```text
Q1 → 25th Percentile
Q2 → 50th Percentile
Q3 → 75th Percentile
```

---


# 📍 First Quartile — Q1

**Q1** is the first quartile.

It represents the **25th percentile**.

Approximately:

> **25% of observations are at or below Q1.**

For example:

```text
Q1 = 25
```

means approximately 25% of the observations are at or below 25.

### 🔑 Remember

```text
Q1 = P25
```

---


# 📍 Second Quartile — Q2

**Q2** is the second quartile.

It represents the **50th percentile**.

The 50th percentile is the **Median**.

Therefore:

```text
Q2 = P50 = Median
```

Approximately 50% of observations are at or below Q2.

### 🔑 Remember

> **Q2 is simply the Median.**

---

# 📍 Third Quartile — Q3

**Q3** is the third quartile.

It represents the **75th percentile**.

Approximately:

> **75% of observations are at or below Q3.**

Therefore:

```text
Q3 = P75
```

---

# 📊 Quartiles at a Glance

| Quartile | Percentile | Meaning |
|---|---:|---|
| Q1 | 25th | 25% of data is at or below this value |
| Q2 | 50th | 50% of data is at or below this value |
| Q3 | 75th | 75% of data is at or below this value |

### 💡 Visualization

```text
Lowest                         Highest
  │                               │
  ▼                               ▼

  0%          25%        50%        75%        100%
  │------------│----------│----------│----------│
               Q1         Q2         Q3
```

---

# 🧮 Example of Quartiles

Consider the following ordered dataset:

```text
10, 20, 30, 40, 50, 60, 70, 80
```

We can use percentile-based calculations to identify:

```text
Q1 → 25th percentile
Q2 → 50th percentile
Q3 → 75th percentile
```

Using Python:

````python
import numpy as np

data = np.array([
    10, 20, 30, 40,
    50, 60, 70, 80
])

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
````

### 💡 Important

Different statistical software can use different percentile interpolation methods.

Therefore, exact percentile values can sometimes differ slightly depending on the method used.

For this repository, we will primarily use the default methods provided by NumPy and Pandas.

# 📏 What is Interquartile Range?

The **Interquartile Range (IQR)** measures the spread of the **middle 50% of the data**.

It ignores the lower 25% and upper 25% when measuring this central spread.

### 🧮 Formula

```text
IQR = Q3 − Q1
```

Where:

- **Q1** = First Quartile
- **Q3** = Third Quartile

### 💡 Simple Meaning

> **IQR tells us how spread out the middle 50% of the data is.**

---

# 🧪 IQR Example

Suppose:

```text
Q1 = 25
Q3 = 75
```

Then:

```text
IQR = Q3 − Q1
    = 75 − 25
    = 50
```

Therefore:

> **IQR = 50**

This means the middle 50% of the data spans a range of 50 units.

---

# 🔍 Why is IQR Useful?

The Range uses:

```text
Minimum → Maximum
```

Therefore, one extreme value can dramatically change the Range.

IQR focuses only on:

```text
Q1 → Q3
```

which represents the central 50% of the data.

### 🔑 Important

> **IQR is much less sensitive to extreme values than Range, Variance, and Standard Deviation.**

This makes IQR particularly useful for:

- 🚨 Outlier detection
- 📊 Skewed distributions
- 🔍 Exploratory Data Analysis
- 📦 Boxplots

---

# ⚖️ Range vs IQR

Consider:

```text
10, 20, 30, 40, 50, 500
```

The Range is:

```text
500 − 10 = 490
```

The value `500` has a huge effect.

IQR focuses on Q1 and Q3 instead.

Therefore, IQR provides a more robust measure of the spread of the central portion of the data.

### Comparison

| 📏 Range | 📦 IQR |
|---|---|
| Uses Minimum and Maximum | Uses Q1 and Q3 |
| Measures total spread | Measures middle 50% spread |
| Highly sensitive to outliers | More resistant to outliers |
| Simple | More useful for robust analysis |

---


# 🚨 IQR and Outlier Detection

One of the most important applications of IQR is **outlier detection**.

A commonly used rule is:

### Lower Bound

```text
Lower Bound = Q1 − 1.5 × IQR
```

### Upper Bound

```text
Upper Bound = Q3 + 1.5 × IQR
```

Observations outside these boundaries are commonly flagged as potential outliers.

### 🔑 Important

> **The 1.5 × IQR rule is a practical statistical rule for identifying potential outliers.**

It does not automatically prove that a value is an error.

An extreme value can be a legitimate observation.

---

# 🧮 Understanding the IQR Outlier Rule

Suppose:

```text
Q1 = 20
Q3 = 60
```

First calculate IQR:

```text
IQR = 60 − 20
    = 40
```

Now calculate:

### Lower Bound

```text
20 − (1.5 × 40)
= 20 − 60
= -40
```

### Upper Bound

```text
60 + (1.5 × 40)
= 60 + 60
= 120
```

Therefore:

```text
Lower Bound = -40
Upper Bound = 120
```

Any observation:

```text
< -40
```

or:

```text
> 120
```

would be flagged as a potential outlier using this rule.

---

# 🐍  Calculating IQR Using Python

Let's calculate IQR using NumPy.

````python
import numpy as np

data = np.array([
    10, 20, 30, 40,
    50, 60, 70, 80
])

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
````

# 🐼 IQR Using Pandas

Pandas provides the `quantile()` method.

````python
import pandas as pd

data = pd.Series([
    10, 20, 30, 40,
    50, 60, 70, 80
])

q1 = data.quantile(0.25)
q3 = data.quantile(0.75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
````

# 🚨  Detecting Outliers Using Python

Let's create a dataset containing an extreme value.

````python
data = np.array([
    10, 20, 25, 30,
    35, 40, 45, 50,
    55, 100
])

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
````

# 🔎 Finding Potential Outliers

We can use Boolean indexing to identify observations outside the IQR boundaries.

````python
outliers = data[
    (data < lower_bound) |
    (data > upper_bound)
]

print("Potential Outliers:", outliers)
````

### 💡 Interpretation

Any value outside the calculated lower and upper bounds is flagged as a potential outlier.

Again:

> **Potential outlier ≠ automatically wrong data**

We should investigate the observation before removing it.

---

# 🧹 Filtering Outliers

If the analysis requires removing observations outside the IQR bounds:

````python
filtered_data = data[
    (data >= lower_bound) &
    (data <= upper_bound)
]

print("Original Data:", data)
print("Filtered Data:", filtered_data)
````

### ⚠️ Important

Outlier removal should not be performed automatically.

Before removing an observation, consider:

- 🌍 Real-world context
- 📊 Data collection process
- 🔍 Whether it is a genuine observation
- 🎯 Business objective
- 🤖 Impact on the ML model

# 📦 What is a Boxplot?

A **Boxplot** is a visualization that summarizes the distribution of numerical data.

It is strongly connected to:

- Q1
- Median
- Q3
- IQR
- Potential outliers

A typical boxplot shows:

```text
Minimum / Lower Whisker
        │
        ├───────────────
        │
       Q1
        │
        │   Box
       Q2  (Median)
        │
       Q3
        │
        ├───────────────
        │
Maximum / Upper Whisker
```

Potential outliers are usually shown as individual points beyond the whiskers.

---

# 📊 Creating a Boxplot

Let's visualize the data.

````python
import matplotlib.pyplot as plt

data = np.array([
    10, 20, 25, 30,
    35, 40, 45, 50,
    55, 100
])

plt.boxplot(data)

plt.title("Boxplot of the Dataset")
plt.ylabel("Values")

plt.show()
````

### 👀 Observation

The boxplot allows us to visually identify:

- Q1
- Median
- Q3
- IQR
- Potential outliers

---

# 📊 Boxplot with Seaborn

Seaborn provides an aesthetically useful interface for boxplots.

````python
import seaborn as sns

sns.boxplot(x=data)

plt.title("Distribution with Boxplot")
plt.xlabel("Values")

plt.show()
````

### 💡 Why Boxplots are Useful

Boxplots are especially useful when comparing distributions across multiple groups.

For example:

```text
Salary by Department
Age by Customer Segment
Sales by Region
Scores by Class
```

# ⚖️ IQR vs Standard Deviation

Both IQR and Standard Deviation measure spread, but they behave differently.

| 📦 IQR | 📏 Standard Deviation |
|---|---|
| Measures middle 50% spread | Measures spread around Mean |
| More resistant to outliers | Sensitive to outliers |
| Uses Q1 and Q3 | Uses Mean |
| Useful for skewed data | Very useful for approximately normal data |
| Used in Boxplots | Used in Z-scores and standardization |

### 🔑 Important

Neither measure is universally better.

The correct choice depends on:

- Distribution
- Outliers
- Data characteristics
- Statistical objective
