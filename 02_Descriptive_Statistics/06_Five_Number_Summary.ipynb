# 📋 Five-Number Summary

A dataset can contain hundreds, thousands, or even millions of observations.

Looking at every individual value is often difficult.

The **Five-Number Summary** provides a compact statistical description of a numerical dataset using five important values:

1. 🔽 Minimum
2. 📦 First Quartile (Q1)
3. 📍 Median (Q2)
4. 📦 Third Quartile (Q3)
5. 🔼 Maximum

Together, these five values help us understand:

- 📍 The center of the data
- 📊 The spread of the data
- 📈 The distribution of the data
- 🚨 Potential outliers
- ⚖️ Whether the data may be skewed

The Five-Number Summary is also the foundation of the **Boxplot**.

# 🧠 What is the Five-Number Summary?

The **Five-Number Summary** describes a dataset using five values:

```text
Minimum
Q1
Median
Q3
Maximum
```

These values divide the ordered data into meaningful sections.

### 📊 Basic Structure

```text
Minimum        Q1          Median          Q3        Maximum
   │            │             │             │            │
   ▼            ▼             ▼             ▼            ▼
   ├────────────┼─────────────┼─────────────┼────────────┤
```

These five values provide a quick overview of the distribution.

---


# 🔽 Minimum

The **Minimum** is the smallest value in the dataset.

### Example

```text
10, 20, 30, 40, 50
```

Minimum:

```text
10
```

Therefore:

> **Minimum = 10**

### 💡 What does it tell us?

The Minimum tells us the lowest observed value in the dataset.

However, if extreme values or data errors are present, the Minimum may not represent the typical lower end of the data.

---


# 📦 First Quartile — Q1

**Q1**, or the First Quartile, represents the **25th percentile**.

```text
Q1 = P25
```

Approximately:

> **25% of observations are at or below Q1.**

### 💡 Example

If:

```text
Q1 = 25
```

then approximately 25% of the observations are at or below 25.

---

# 📍 Median — Q2

The **Median** is the middle value of an ordered dataset.

It is also the:

```text
50th Percentile
```

and:

```text
Q2 = P50
```

Therefore:

> **Q2 = Median**

Approximately 50% of observations are at or below the Median.

---


# 📦 Third Quartile — Q3

**Q3**, or the Third Quartile, represents the **75th percentile**.

```text
Q3 = P75
```

Approximately:

> **75% of observations are at or below Q3.**

---

# 🔼 Maximum

The **Maximum** is the largest observed value in the dataset.

### Example

```text
10, 20, 30, 40, 50
```

Maximum:

```text
50
```

Therefore:

> **Maximum = 50**

Like Minimum, Maximum can be strongly affected by extreme observations.

---

# 📊  The Five Values Together

The Five-Number Summary can be represented as:

```text
Minimum → Q1 → Median → Q3 → Maximum
```

Or:

```text
Min ─── Q1 ─── Median ─── Q3 ─── Max
```

### Relationship with Percentiles

| Five-Number Value | Percentile |
|---|---:|
| Minimum | 0th percentile* |
| Q1 | 25th percentile |
| Median | 50th percentile |
| Q3 | 75th percentile |
| Maximum | 100th percentile* |

\*Conceptually, Minimum and Maximum correspond to the endpoints of the observed dataset, though percentile conventions can vary.

# 🐍  Five-Number Summary Using NumPy

Let's calculate the five values using NumPy.

````python
import numpy as np

data = np.array([
    10, 20, 30, 40,
    50, 60, 70, 80, 90
])

minimum = np.min(data)
q1 = np.percentile(data, 25)
median = np.percentile(data, 50)
q3 = np.percentile(data, 75)
maximum = np.max(data)

print("Minimum:", minimum)
print("Q1:", q1)
print("Median:", median)
print("Q3:", q3)
print("Maximum:", maximum)
````

# 📋 Creating a Five-Number Summary Dictionary

We can organize the results into a dictionary.

````python
five_number_summary = {
    "Minimum": minimum,
    "Q1": q1,
    "Median": median,
    "Q3": q3,
    "Maximum": maximum
}

five_number_summary
````

# 🐼  Five-Number Summary Using Pandas

Pandas provides the `quantile()` method.

````python
import pandas as pd

data = pd.Series([
    10, 20, 30, 40,
    50, 60, 70, 80, 90
])

five_number_summary = pd.Series({
    "Minimum": data.min(),
    "Q1": data.quantile(0.25),
    "Median": data.median(),
    "Q3": data.quantile(0.75),
    "Maximum": data.max()
})

five_number_summary
````

# 📏 Interquartile Range from Five-Number Summary

The Five-Number Summary allows us to calculate the IQR immediately.

### Formula

```text
IQR = Q3 − Q1
```

Suppose:

```text
Q1 = 30
Q3 = 70
```

Then:

```text
IQR = 70 − 30
    = 40
```

Therefore:

> **IQR = 40**

The IQR represents the spread of the **middle 50%** of the observations.

---

# 🚨 Outlier Detection Using Five-Number Summary

The Five-Number Summary provides Q1 and Q3.

These can be used to calculate IQR.

Then we can use the common **1.5 × IQR rule**.

### Step 1 — Calculate IQR

```text
IQR = Q3 − Q1
```

### Step 2 — Calculate Lower Bound

```text
Lower Bound = Q1 − 1.5 × IQR
```

### Step 3 — Calculate Upper Bound

```text
Upper Bound = Q3 + 1.5 × IQR
```

### Step 4 — Identify Potential Outliers

Values below the Lower Bound or above the Upper Bound can be flagged as potential outliers.

---


# 🧪 Outlier Detection Using Python

Let's use a dataset containing an extreme value.


````python
data = np.array([
    10, 12, 15, 18,
    20, 22, 25, 28,
    30, 32, 35, 38,
    40, 42, 100
])

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = data[
    (data < lower_bound) |
    (data > upper_bound)
]

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Potential Outliers:", outliers)
````

### 💡 Interpretation

The value `100` may be flagged as a potential outlier if it falls beyond the upper IQR boundary.

But remember:

> **A statistical outlier is not automatically an incorrect observation.**

# 📈 Five-Number Summary and Skewness

The relative spacing between the five values can provide clues about the shape of a distribution.

Consider:

```text
Min    Q1    Median    Q3    Max
│──────│──────│────────│──────│
```

If the distances between the values are roughly balanced, the distribution may be relatively symmetric.

If one side has a much longer tail, the distribution may be skewed.

### ➡️ Right-Skewed Example

```text
Min ─ Q1 ─ Median ─ Q3 ───────── Max
```

The upper side is more spread out.

### ⬅️ Left-Skewed Example

```text
Min ───────── Q1 ─ Median ─ Q3 ─ Max
```

The lower side is more spread out.

### ⚠️ Important

The Five-Number Summary can provide clues about skewness, but a complete conclusion should consider:

- Distribution plots
- Histogram
- Boxplot
- Mean vs Median
- Skewness statistics

# 🧠  Five-Number Summary vs Mean and Standard Deviation

The Five-Number Summary is based on:

```text
Minimum
Q1
Median
Q3
Maximum
```

Mean and Standard Deviation use:

```text
Mean
Variance
Standard Deviation
```

### Comparison

| 📋 Five-Number Summary | 📊 Mean + Standard Deviation |
|---|---|
| Uses Median and Quartiles | Uses Mean |
| More robust to extreme values | More sensitive to outliers |
| Useful for skewed data | Useful for many approximately symmetric distributions |
| Basis of Boxplot | Useful for Z-scores |
| Strong EDA tool | Strong statistical and ML tool |

### 💡 Important

Both approaches are useful.

A good Data Scientist understands when to use each.

# 🧮 Building a Reusable Function

We can create a simple Python function to calculate the Five-Number Summary.

````python
def five_number_summary(data):
    return {
        "Minimum": np.min(data),
        "Q1": np.percentile(data, 25),
        "Median": np.percentile(data, 50),
        "Q3": np.percentile(data, 75),
        "Maximum": np.max(data)
    }
````

````python
data = np.array([
    10, 20, 30, 40,
    50, 60, 70, 80, 90
])

summary = five_number_summary(data)

summary
````


### 💡 Why Functions?

Reusable functions help us:

- ♻️ Avoid repeated code
- 🧹 Keep notebooks clean
- 📊 Analyze multiple datasets
- 🛠️ Build practical data-analysis workflows

---


# 📊 Applying the Function to Multiple Features

Suppose we have:

````python
df = pd.DataFrame({
    "Age": [20, 22, 24, 26, 28, 30],
    "Salary": [25000, 30000, 35000, 40000, 45000, 50000],
    "Experience": [1, 2, 3, 4, 5, 6]
})

df
````

````python
# We can calculate the Five-Number Summary for each numerical feature:

for column in df.columns:
    print(f"\n{column}")
    print(five_number_summary(df[column].values))
````
