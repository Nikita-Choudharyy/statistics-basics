# 🧠 Why Do We Need Measures of Spread?

Suppose two students have the following marks across five tests.

### Student A

```text

```

### Student B

```text

```


### Student A

### Student B

### 🔑 Important

# 📐  What is Variance?

**Variance** measures how much the observations differ from the mean, on average, using **squared deviations**.

In simple words:

> **Variance tells us how spread out the data is around its mean.**

A small variance means observations are relatively close to the mean.

A large variance means observations are more spread out.

---

# 💡 Intuition Behind Variance

Consider:

```text
10, 20, 30, 40, 50
```

The mean is:

```text
30
```

Now calculate how far each value is from the mean:

| Value | Mean | Deviation |
|---:|---:|---:|
| 10 | 30 | -20 |
| 20 | 30 | -10 |
| 30 | 30 | 0 |
| 40 | 30 | +10 |
| 50 | 30 | +20 |

If we simply add these deviations:

```text
-20 + (-10) + 0 + 10 + 20 = 0
```

The positive and negative deviations cancel each other.

So we square the deviations.

```text
(-20)² = 400
(-10)² = 100
(0)² = 0
(10)² = 100
(20)² = 400
```

Now:

```text
400 + 100 + 0 + 100 + 400 = 1000
```

Variance is based on the average of these squared deviations.

### 🔑 Core Idea

```text
Distance from Mean
        ↓
Deviation
        ↓
Square the Deviation
        ↓
Average the Squared Deviations
        ↓
Variance
```

---

# 🧮  Population Variance Formula

For an entire population, Variance is represented by:

```text
σ²
```

The formula is:

```text
σ² = Σ(Xᵢ − μ)² / N
```

Where:

- **σ²** = Population Variance
- **Xᵢ** = Individual population observation
- **μ** = Population Mean
- **N** = Population size
- **Σ** = Sum of all squared deviations

### 💡 In Simple Words

```text
Population Variance
=
Average of squared distances from the population mean
```
 ---

# 🧪 Manual Calculation of Population Variance

Consider:

```text
Data = [10, 20, 30, 40, 50]
```

### Step 1 — Calculate Mean

```text
Mean = 30
```

### Step 2 — Calculate Deviations

```text
10 − 30 = -20
20 − 30 = -10
30 − 30 = 0
40 − 30 = 10
50 − 30 = 20
```

### Step 3 — Square Deviations

```text
400
100
0
100
400
```

### Step 4 — Add Squared Deviations

```text
400 + 100 + 0 + 100 + 400 = 1000
```

### Step 5 — Divide by Number of Observations

```text
Variance = 1000 / 5
         = 200
```

Therefore:

> **Population Variance = 200**

---

# 📏  What is Standard Deviation?

**Standard Deviation** is the square root of Variance.

It tells us how much the observations typically vary around the mean, in the **same units as the original data**.

### Formula

For a population:

```text
σ = √σ²
```

Since:

```text
Variance = 200
```

we get:

```text
Standard Deviation = √200
                   ≈ 14.14
```

Therefore:

> **Population Standard Deviation ≈ 14.14**

### 🔑 Important

> **Variance is measured in squared units, while Standard Deviation is measured in the original units.**

For example:

If the data represents salary in rupees:

```text
Standard Deviation → Rupees
Variance → Rupees²
```

This is one reason Standard Deviation is often easier to interpret.

---

# 🧮 7. Standard Deviation Formula

For a population:

```text
σ = √[Σ(Xᵢ − μ)² / N]
```

Where:

- **σ** = Population Standard Deviation
- **Xᵢ** = Individual observation
- **μ** = Population Mean
- **N** = Population size

### 💡 In Simple Words

```text
Standard Deviation
=
Square root of the average squared distance from the mean
```
 ---

# ⚖️ 8. Variance vs Standard Deviation

| 📐 Variance | 📏 Standard Deviation |
|---|---|
| Measures spread | Measures spread |
| Uses squared deviations | Square root of variance |
| Units are squared | Same units as original data |
| Harder to interpret directly | Easier to interpret |
| Useful in mathematical/statistical calculations | Useful for practical interpretation |

### 🔑 Remember

```text
Variance → Squared Spread

Standard Deviation → Spread in Original Units
```

---

# 👥  Population Variance vs Sample Variance

There are two important cases.

### 📊 Population Variance

Used when we have the **entire population**.

Formula:

```text
σ² = Σ(Xᵢ − μ)² / N
```

We divide by:

```text
N
```

---

### 🎯 Sample Variance

Used when we have only a **sample** from a population.

Formula:

```text
s² = Σ(xᵢ − x̄)² / (n − 1)
```

Where:

- **s²** = Sample Variance
- **xᵢ** = Sample observation
- **x̄** = Sample Mean
- **n** = Sample size

We divide by:

```text
n − 1
```

This is known as **Bessel's correction**.

### 💡 Why `n − 1`?

When estimating population variability from a sample, dividing by `n` tends to underestimate the population variance on average.

Using `n − 1` corrects for this bias under the standard assumptions.

 ---

# 🧠 Why Do We Square Deviations?

You may wonder:

> **Why don't we simply take the average of the absolute deviations?**

The main reason for squaring is that positive and negative deviations would otherwise cancel.

For example:

```text
-20 + (-10) + 0 + 10 + 20 = 0
```

Squaring makes all deviations non-negative:

```text
400 + 100 + 0 + 100 + 400
```

This allows us to measure the total amount of variation around the mean.

Squaring also gives Variance useful mathematical properties that are important in probability, statistics, and Machine Learning.

---

# 🐍 Variance Using NumPy

Let's calculate population variance using NumPy.

````python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

variance = np.var(data)

print("Population Variance:", variance)
````

# 📏 Standard Deviation Using NumPy

NumPy provides `np.std()`.

````python
standard_deviation = np.std(data)

print("Population Standard Deviation:", standard_deviation)
````

````python
#We can round the result:

print("Standard Deviation:", round(standard_deviation, 2))
````

# 🎯 Sample Variance Using NumPy

By default:

```python
np.var()
```

calculates population variance using:

```text
ddof = 0
```

For sample variance, use:

````python
sample_variance = np.var(data, ddof=1)

print("Sample Variance:", sample_variance)
````

# 🎯 Sample Standard Deviation Using NumPy

For sample standard deviation:

````python
sample_std = np.std(data, ddof=1)

print("Sample Standard Deviation:", sample_std)

#We can round the result:
print("Sample Standard Deviation:", round(sample_std, 2))
````

# 🐼 Variance Using Pandas

Pandas uses sample variance by default for a Series.

````python
import pandas as pd

data = pd.Series([10, 20, 30, 40, 50])

variance = data.var()

print("Sample Variance:", variance)
````

# 📏 Standard Deviation Using Pandas

````python
standard_deviation = data.std()

print("Sample Standard Deviation:", standard_deviation)
````

### 🔑 Important

Pandas:

```python
data.var()
data.std()
```

use sample statistics by default (`ddof=1`).

NumPy:

```python
np.var(data)
np.std(data)
```

use population calculations by default (`ddof=0`).

Always check the convention being used when comparing results.

# 📊 Comparing Low and High Variance

Let's create two datasets.

````python
data_low_variation = np.array([48, 49, 50, 51, 52])

data_high_variation = np.array([10, 30, 50, 70, 90])

print(
    "Low Variation - Standard Deviation:",
    np.std(data_low_variation)
)

print(
    "High Variation - Standard Deviation:",
    np.std(data_high_variation)
)
````

### 👀 Observation

The first dataset has values close to its mean.

Therefore:

```text
Lower Standard Deviation
```

The second dataset has values spread far away from its mean.

Therefore:

```text
Higher Standard Deviation
```

### 🔑 Core Idea

> **Small Standard Deviation → Data is concentrated around the mean.**

> **Large Standard Deviation → Data is more spread out around the mean.**

---

# 📈 Visualizing Standard Deviation

Visualization helps us understand the concept more intuitively.

````python
import matplotlib.pyplot as plt

data = np.array([48, 49, 50, 51, 52])

mean = np.mean(data)
std = np.std(data)

plt.hist(data, bins=5, edgecolor="black")

plt.axvline(
    mean,
    linestyle="--",
    label=f"Mean = {mean}"
)

plt.axvline(
    mean + std,
    linestyle=":",
    label=f"Mean + 1 SD = {mean + std:.2f}"
)

plt.axvline(
    mean - std,
    linestyle=":",
    label=f"Mean - 1 SD = {mean - std:.2f}"
)

plt.title("Mean and Standard Deviation")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.legend()

plt.show()
````

### 👀 Observation

The Standard Deviation gives us a way to describe the typical spread around the mean.

# 📊 19. Standard Deviation and Normal Distribution

Standard Deviation becomes especially important when data follows approximately a **Normal Distribution**.

For a normal distribution:

```text
Mean ± 1 Standard Deviation
```

contains approximately:

```text
68% of observations
```

Similarly:

```text
Mean ± 2 Standard Deviations
```

contains approximately:

```text
95% of observations
```

And:

```text
Mean ± 3 Standard Deviations
```

contains approximately:

```text
99.7% of observations
```

This is known as the:

> **68–95–99.7 Rule**

### ⚠️ Important

This rule applies to approximately **normal distributions**.

It should not automatically be applied to every dataset.

# 🚨 Effect of Outliers on Variance

Variance is sensitive to extreme values because deviations are **squared**.

Consider:

```text
10, 20, 30, 40, 50
```

Now add:

```text
500
```

The deviation of 500 from the mean is very large.

When squared, that deviation becomes even larger.

Therefore, an outlier can significantly increase:

- Variance
- Standard Deviation

### 🔑 Important

> **Because Variance squares deviations, extreme observations can have a strong influence on it.**

---

# 🧪 Demonstrating the Effect of an Outlier

````python
data_without_outlier = np.array([10, 20, 30, 40, 50])

data_with_outlier = np.array([10, 20, 30, 40, 500])

print("Without Outlier")
print("Variance:", np.var(data_without_outlier))
print("Standard Deviation:", np.std(data_without_outlier))

print("\nWith Outlier")
print("Variance:", np.var(data_with_outlier))
print("Standard Deviation:", np.std(data_with_outlier))
````

### 👀 Observation

The extreme value causes a large increase in both:

```text
Variance
```

and:

```text
Standard Deviation
```

This is why understanding outliers is important before interpreting these measures.

# 📏 Standard Deviation and Feature Scaling

Standard Deviation is also important in **Standardization**.

A common standardization formula is:

```text
z = (x − μ) / σ
```

Where:

- **x** = Original value
- **μ** = Mean
- **σ** = Standard Deviation
- **z** = Standardized value

This transformation produces a feature with approximately:

```text
Mean = 0
Standard Deviation = 1
```

when applied in the standard way.

This is commonly called:

> **Z-score Standardization**

 ---


# 🤖 Why Standardization Matters in Machine Learning

Some Machine Learning algorithms are sensitive to feature scale.

Suppose we have:

```text
Age:
20–60

Income:
20,000–500,000
```

Income has much larger numerical values.

Standardization can transform features into a comparable scale.

It is commonly useful for algorithms such as:

- 📍 K-Nearest Neighbors
- 📐 Logistic Regression
- 📈 Linear Regression
- 🎯 Support Vector Machines
- 🧠 Neural Networks
- 📊 Principal Component Analysis

### ⚠️ Important

Not every Machine Learning algorithm requires standardization.

For example, tree-based algorithms are generally much less sensitive to feature scaling.

---

# 🧪 Standardization Using Scikit-Learn

````python
from sklearn.preprocessing import StandardScaler

data = np.array([
    [20],
    [25],
    [30],
    [35],
    [40]
])

scaler = StandardScaler()

scaled_data = scaler.fit_transform(data)

print(scaled_data)
````

The transformed values are centered around:

```text
Mean ≈ 0
```

and have:

```text
Standard Deviation ≈ 1
```

### 🔑 Connection

Standardization directly relies on:

- Mean
- Standard Deviation

Therefore, understanding these concepts is essential before studying feature scaling in Machine Learning.

# 🤖 Variance in Machine Learning

Variance appears in several Machine Learning concepts.

### 📊 Feature Analysis

Variance tells us how much a feature varies.

### 🧹 Data Preprocessing

Very low-variance features may contain little useful information.

This leads to the idea of:

> **Variance Threshold**

where features with extremely low variance can sometimes be removed.

### 🎯 Bias-Variance Tradeoff

Variance is also part of the famous:

> **Bias-Variance Tradeoff**

High model variance can indicate that a model is too sensitive to training data and may overfit.

This is a different but related use of the word "variance."

### ⚠️ Important

> **Statistical variance of a feature and model variance in the bias-variance tradeoff are related through variability, but they are not the same concept.**
