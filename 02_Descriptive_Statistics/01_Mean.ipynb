## 🧠  What is Mean?

The **Mean** is a measure of central tendency that represents the average value of a dataset.

The basic idea is:

> **Add all values together and divide the total by the number of values.**

### 💡 Simple Example

Suppose the marks of five students are:

```text
60, 70, 80, 90, 100
```

First, add all the values:

```text
60 + 70 + 80 + 90 + 100 = 400
```

There are 5 observations.

Therefore:

```text
Mean = 400 / 5
     = 80
```

So, the mean marks are:

**80**

### 🔑 Intuition

The mean can be thought of as the **equal-sharing point** of the data.

If the total value were distributed equally among all observations, each observation would receive the mean.

---



## 🧮 Mathematical Formula

For a dataset containing:

```text
x₁, x₂, x₃, ..., xₙ
```

the arithmetic mean is:

```text
Mean = (x₁ + x₂ + x₃ + ... + xₙ) / n
```

It can also be written as:

```text
x̄ = Σxᵢ / n
```

Where:

- **x̄** = Sample Mean
- **Σxᵢ** = Sum of all observations
- **n** = Number of observations

### 💡 In Simple Words

```text
Mean = Sum of all values / Number of values
```

---


## 📐  Population Mean

When we calculate the mean using **every observation in a population**, it is called the **Population Mean**.

The population mean is represented by:

**μ (mu)**

### Formula

```text
μ = ΣXᵢ / N
```

Where:

- **μ** = Population Mean
- **ΣXᵢ** = Sum of all population values
- **N** = Population size

### Example

Suppose the complete population is:

```text
10, 20, 30, 40, 50
```

Then:

```text
μ = (10 + 20 + 30 + 40 + 50) / 5
μ = 150 / 5
μ = 30
```

Therefore:

**Population Mean = 30**

---



## 🎯  Sample Mean

When we calculate the mean using a **sample** taken from a population, it is called the **Sample Mean**.

The sample mean is represented by:

**x̄ (x-bar)**

### Formula

```text
x̄ = Σxᵢ / n
```

Where:

- **x̄** = Sample Mean
- **Σxᵢ** = Sum of sample observations
- **n** = Sample size

### 🔑 Important

> **Population → μ**

> **Sample → x̄**

We usually use the sample mean to estimate the population mean when the entire population is not available.

---


## ⚖️  Population Mean vs Sample Mean

| 📐 Population Mean | 🎯 Sample Mean |
|---|---|
| Calculated from the entire population | Calculated from a sample |
| Symbol: **μ** | Symbol: **x̄** |
| Population size = **N** | Sample size = **n** |
| Describes the population | Describes the sample |
| Usually a parameter | Usually a statistic |

### 💡 Remember

**Population → Parameter → μ**

**Sample → Statistic → x̄**

## 🧪  Manual Calculation

Let's calculate the mean manually.

Suppose we have:

```text
Data = [10, 20, 30, 40, 50]
```

### Step 1 — Add the values

```text
10 + 20 + 30 + 40 + 50 = 150
```

### Step 2 — Count the observations

```text
n = 5
```

### Step 3 — Divide

```text
Mean = 150 / 5
     = 30
```

Therefore:

> **Mean = 30**

 --- 

## 🐍  Calculating Mean Using Python

Python provides several ways to calculate the mean.

We will start with **NumPy**.

````python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

mean = np.mean(data)

print("Mean:", mean)
````



## 🔍  Understanding the Result

Our dataset is:

```text
[10, 20, 30, 40, 50]
```

The calculated mean is:

```text
30
```

This means that if the total value of the dataset were distributed equally among all five observations, each observation would receive **30**.

Notice that **30 does not need to be the most frequently occurring value**.

It simply represents the arithmetic center of the data.

---

## 🐼  Mean Using Pandas

The mean can also be calculated using Pandas.

````python
import pandas as pd

data = pd.Series([10, 20, 30, 40, 50])

mean = data.mean()

print("Mean:", mean)
````

## 📈  Visualizing the Mean

Visualization can help us understand where the mean lies within a dataset.

````python
import numpy as np
import matplotlib.pyplot as plt

data = np.array([10, 20, 30, 40, 50])

mean = np.mean(data)

plt.hist(data, bins=5, edgecolor="black")

plt.axvline(
    mean,
    linestyle="--",
    label=f"Mean = {mean}"
)

plt.title("Mean of the Dataset")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.legend()

plt.show()
````


### 👀 Observation

The vertical line represents the mean of the dataset.

For a simple symmetric dataset like this one, the mean lies near the center of the observations.

## ⚠️ Mean and Outliers

One important property of the mean is that it is **sensitive to extreme values**, also called outliers.

Consider this dataset:

```text
10, 20, 30, 40, 50
```

Its mean is:

```text
30
```

Now suppose an extreme value is added:

```text
10, 20, 30, 40, 500
```

The mean becomes:

```text
120
```

The extreme value **500** has significantly changed the mean.

### 🔑 Important

> **The mean is sensitive to outliers.**

This becomes important when analyzing real-world datasets.

---


## 🧪  Demonstrating the Effect of an Outlier

````python
import numpy as np 

normal_data = np.array([10, 20, 30, 40, 50])
data_with_outlier = np.array([10, 20, 30, 40, 500])

mean_normal = np.mean(normal_data)
mean_outlier = np.mean(data_with_outlier)

print("Mean without outlier:", mean_normal)
print("Mean with outlier:", mean_outlier)
````

### 💡 Interpretation

A single extreme value changed the mean from:

```text
30 → 120
```

This does not mean that the mean is a bad measure.

It means we should understand the **distribution and presence of outliers** before interpreting the mean.

Later, when we study **Median and Mode**, we will compare how they behave in the presence of outliers.

## ⚖️  Weighted Mean

Sometimes every observation does not have the same importance.

In such situations, we can use a **Weighted Mean**.

The weighted mean gives different observations different levels of importance.

### Formula

```text
Weighted Mean = Σ(wᵢxᵢ) / Σwᵢ
```

Where:

- **xᵢ** = Observation
- **wᵢ** = Weight assigned to the observation
- **Σ(wᵢxᵢ)** = Sum of weighted values
- **Σwᵢ** = Sum of all weights

### 💡 Example

Suppose a student's final grade is calculated using:

| Component | Score | Weight |
|---|---:|---:|
| Assignment | 80 | 20% |
| Midterm | 70 | 30% |
| Final Exam | 90 | 50% |

The weighted mean is:

```text
(80 × 0.20) + (70 × 0.30) + (90 × 0.50)
```

```text
= 16 + 21 + 45
= 82
```

Therefore:

**Weighted Mean = 82**

### 🐍  Weighted Mean Using Python

We can calculate the weighted mean using NumPy.

````python
import numpy as np

scores = np.array([80, 70, 90])
weights = np.array([0.20, 0.30, 0.50])

weighted_mean = np.average(
    scores,
    weights=weights
)

print("Weighted Mean:", weighted_mean)
````


### 💡 Real-World Use

Weighted averages are commonly used in:

- 🎓 Academic grading
- 💰 Financial calculations
- 📊 Survey analysis
- 📈 Performance scoring
- 🤖 Feature weighting in some analytical methods

## ⚠️  When Should We Be Careful with Mean?

Mean should be interpreted carefully when:

- 🚨 Strong outliers are present
- 📈 Data is heavily skewed
- 📊 The distribution is highly asymmetric
- 🔍 The mean does not represent a typical observation

For example, consider salaries:

```text
₹20,000
₹22,000
₹25,000
₹28,000
₹5,00,000
```

The very high salary can pull the mean upward.

In such situations, the **median** may provide a better representation of the typical value.
