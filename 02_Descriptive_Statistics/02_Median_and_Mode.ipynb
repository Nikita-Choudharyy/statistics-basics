# 📍 What is Median?

The **Median** is the middle value of a dataset when the observations are arranged in **ascending or descending order**.

### 💡 Simple Example

Consider:

```text
10, 20, 30, 40, 50
```

The values are already sorted.

The middle value is:

```text
30
```

Therefore:

> **Median = 30**

The Median divides the ordered dataset into two parts:

- Approximately 50% of observations are below it.
- Approximately 50% of observations are above it.

### 🔑 Key Idea

> **Median represents the middle position of ordered data.**

---


# 🔢  Median When Number of Observations is Odd

When the number of observations is **odd**, there is exactly one middle value.

For example:

```text
10, 20, 30, 40, 50
```

There are:

```text
n = 5
```

The middle position is:

```text
(n + 1) / 2
```

```text
(5 + 1) / 2 = 3
```

Therefore, the third value is the Median:

```text
Median = 30
```

### 📌 Formula

For an odd number of observations:

```text
Median position = (n + 1) / 2
```

---

# 🔢  Median When Number of Observations is Even

When the number of observations is **even**, there are two middle values.

Consider:

```text
10, 20, 30, 40, 50, 60
```

There are:

```text
n = 6
```

The two middle values are:

```text
30 and 40
```

We calculate their average:

```text
Median = (30 + 40) / 2
       = 35
```

Therefore:

> **Median = 35**

### 🔑 Important

For an even number of observations:

```text
Median = Average of the two middle values
```

---

# 🧮  Mathematical Intuition of Median

Suppose the sorted observations are:

```text
x₁, x₂, x₃, ..., xₙ
```

For an odd number of observations, the Median is the middle observation.

For an even number of observations, the Median is the average of the two middle observations.

The Median depends primarily on the **position of observations after sorting**, rather than the magnitude of every individual observation.

This is why Median is generally less affected by extreme values than Mean.

---

# 🧪  Manual Calculation — Odd Dataset

Let's calculate the Median manually.

```text
Data = [12, 18, 25, 30, 42]
```

### Step 1 — Sort the data

```text
12, 18, 25, 30, 42
```

### Step 2 — Find the middle value

```text
25
```

Therefore:

```text
Median = 25
```

---

# 🧪  Manual Calculation — Even Dataset

Now consider:

```text
Data = [10, 20, 30, 40, 50, 60]
```

The two middle values are:

```text
30 and 40
```

Therefore:

```text
Median = (30 + 40) / 2
       = 35
```

So:

```text
Median = 35
```

---

# 🐍 Median Using NumPy

We can calculate the Median easily using NumPy.


````python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

median = np.median(data)

print("Median:", median)
````

## 🧮  Median of an Even Dataset

````python
data = np.array([10, 20, 30, 40, 50, 60])

median = np.median(data)

print("Median:", median)
````

# 🐼  Median Using Pandas

Pandas also provides a convenient `median()` method.

````python
import pandas as pd

data = pd.Series([10, 20, 30, 40, 50])

median = data.median()

print("Median:", median)
````

# 🔢  What is Mode?

The **Mode** is the value that appears **most frequently** in a dataset.

### 💡 Example

Consider:

```text
10, 20, 20, 30, 40
```

The value:

```text
20
```

appears twice.

The other values appear only once.

Therefore:

> **Mode = 20**

### 🔑 Key Idea

> **Mode is the most frequently occurring value.**

---


# 🧠  Why is Mode Useful?

Mode is especially useful when we want to identify the **most common observation or category**.

For example:

### 👕 Clothing Store

```text
Small
Medium
Medium
Large
Medium
Small
```

The most common size is:

```text
Medium
```

Therefore:

> **Mode = Medium**

This is useful even though clothing sizes are categorical data.

### 🔑 Important

Unlike Mean and Median, Mode can be used with **categorical data**.

---

# 🧩  Types of Mode

A dataset can have different numbers of modes.

### 1️⃣ Unimodal

One mode.

Example:

```text
10, 20, 20, 30, 40
```

Mode:

```text
20
```



### 2️⃣ Bimodal

Two modes.

Example:

```text
10, 10, 20, 30, 30
```

Modes:

```text
10 and 30
```



### 3️⃣ Multimodal

More than two modes.

Example:

```text
10, 10, 20, 20, 30, 30
```

Modes:

```text
10, 20, 30
```



### 4️⃣ No Mode

If every value occurs only once, there may be no mode.

Example:

```text
10, 20, 30, 40, 50
```

No value occurs more frequently than another.

Therefore:

> **No unique mode exists.**

---


# 🐍  Mode Using Pandas

Pandas provides the `mode()` method.

````python
data = pd.Series([10, 20, 20, 30, 40])

mode = data.mode()

print("Mode:")
print(mode)
````

# 📊  Mode Using SciPy

SciPy also provides a statistical mode function.

````python
from scipy import stats

data = [10, 20, 20, 30, 40]

mode_result = stats.mode(data, keepdims=True)

print("Mode:", mode_result.mode[0])
````

# ⚖️  Mean vs Median vs Mode

The three measures of central tendency provide different information.

| 📊 Measure | Meaning | Sensitive to Outliers? | Suitable for Categorical Data? |
|---|---|---:|---:|
| Mean | Arithmetic average | ✅ Yes | ❌ No |
| Median | Middle value | 🟢 Less sensitive | ❌ Generally no |
| Mode | Most frequent value | 🟢 Usually not | ✅ Yes |

### 💡 Simple Memory Trick

**Mean → Average**

**Median → Middle**

**Mode → Most frequent**

---

# 🚨  Effect of Outliers

One of the most important differences between Mean and Median is their response to outliers.

Consider:

```text
10, 20, 30, 40, 50
```

Mean:

```text
30
```

Median:

```text
30
```

Now add an extreme value:

```text
10, 20, 30, 40, 500
```

Mean becomes:

```text
120
```

But Median remains:

```text
30
```

This demonstrates an important property:

> **Median is more resistant to extreme values than Mean.**




---

# 🧪  Demonstrating Outlier Effect Using Python

````python
data_without_outlier = np.array([10, 20, 30, 40, 50])
data_with_outlier = np.array([10, 20, 30, 40, 500])

mean_without = np.mean(data_without_outlier)
median_without = np.median(data_without_outlier)

mean_with = np.mean(data_with_outlier)
median_with = np.median(data_with_outlier)

print("Without Outlier")
print("Mean:", mean_without)
print("Median:", median_without)

print("\nWith Outlier")
print("Mean:", mean_with)
print("Median:", median_with)
````

### 👀 Observation

The outlier changed:

```text
Mean: 30 → 120
```

But:

```text
Median: 30 → 30
```

This shows why Median can be more useful for datasets containing extreme values.

---

# 📈 Mean, Median and Skewness

The relationship between Mean and Median can provide clues about the shape of a distribution.

### 😊 Symmetric Distribution

In a perfectly symmetric distribution:

```text
Mean ≈ Median ≈ Mode
```

---

### ➡️ Right-Skewed Distribution

In a positively skewed distribution, the long tail extends toward larger values.

A common relationship is:

```text
Mean > Median
```

---

### ⬅️ Left-Skewed Distribution

In a negatively skewed distribution, the long tail extends toward smaller values.

A common relationship is:

```text
Mean < Median
```

### 🔑 Important

These relationships are useful as general intuition, but they are not universal rules for every dataset.

---


# 📊  Visualizing Mean and Median

Let's create a dataset and visualize both.

````python
import matplotlib.pyplot as plt

data = np.array([10, 20, 30, 40, 50, 500])

mean = np.mean(data)
median = np.median(data)

plt.hist(data, bins=6, edgecolor="black")

plt.axvline(
    mean,
    linestyle="--",
    label=f"Mean = {mean:.1f}"
)

plt.axvline(
    median,
    linestyle=":",
    label=f"Median = {median:.1f}"
)

plt.title("Mean vs Median with an Outlier")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()

plt.show()
````

### 👀 Observation

The extreme value pulls the Mean toward the right.

The Median remains closer to the center of the majority of observations.

This is why Median can often be a better measure of the "typical" value for heavily skewed data.

# 🌍  Real-World Example — Income

Income is a common example where Mean and Median can tell very different stories.

Suppose five people earn:

```text
₹20,000
₹22,000
₹25,000
₹28,000
₹5,00,000
```

The very high income strongly affects the Mean.

The Median remains closer to the income of the majority of people.

### 💡 Why This Matters

When analyzing income data:

- Mean tells us the arithmetic average.
- Median tells us the middle income.
- The difference between them can indicate the influence of high incomes.

Therefore, analysts often report both Mean and Median.

# 🤖  Median in Data Science

Median is widely used in Exploratory Data Analysis.

It is especially useful when:

- Data is skewed
- Outliers are present
- We want a robust measure of central tendency
- Income or transaction data contains extreme values

For example:

```python
df["Salary"].median()
```

can provide a more representative central value than the mean when salaries are highly skewed.

---


# 🤖  Mode in Data Science

Mode is particularly useful for categorical variables.

Examples:

- Most common city
- Most common product category
- Most common customer segment
- Most common device type
- Most common response category

Using Pandas:

```python
df["City"].mode()
```

can identify the most frequently occurring city.

---

# 🧹  Median and Missing Values

Median can also be used for missing-value imputation.

Suppose:

```text
Age:
20
22
25
NaN
28
100
```

The value `100` may be an extreme observation.

Using the mean could pull the replacement value upward.

In some situations, using the median can provide a more robust replacement.

Example:

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

### ⚠️ Important

Median imputation is not automatically the best method.

The appropriate missing-value strategy depends on:

- Data distribution
- Missing-data mechanism
- Feature meaning
- Outliers
- Modeling objective
